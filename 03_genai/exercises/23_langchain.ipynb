# 🔗 **What is LangChain?**

# **LangChain = Framework / Ecosystem for Building LLM-Powered Applications**

LangChain is **not an LLM**.

**It is an application framework that provides reusable abstractions for connecting:**

```text
Models
   +
Prompts
   +
Messages
   +
Chains
   +
Retrievers
   +
Tools
   +
Output Parsers
   +
Agents
   +
Workflows
```

**The main idea is:**

> **Instead of manually connecting every LLM application component, LangChain provides standard interfaces and reusable building blocks.**

**A useful mental model is:**

```text
Your Application
       ↓
   LangChain
       ↓
 ┌─────┼─────────┬─────────┐
 ↓     ↓         ↓         ↓
Model Prompt   Tools     Retrieval
       ↓
   Application
```

> **LangChain is an application framework around models, not the model itself.**

# 1️⃣ **LLMs**

A traditional LLM abstraction receives text and returns text.

```text

 ↓

 ↓

 ↓

```

### **Examples**

```text

```

**Conceptually:**

```text

              ↓

```

---

# 2️⃣ **Chat Models**

> **Chat Model = A model interface designed around structured messages and conversational interactions.**

Instead of treating the entire input as one plain string, chat models work with messages.

```text
Input
 ↓
List of Messages
 ↓
Chat Model
 ↓
AI Message
```

### **Typical use cases**

```text
Chatbots
Question Answering
Assistants
Multi-turn conversations
Tool Calling
Instruction Following
```

---

# 🆚 **LLM vs Chat Model**

| Feature | LLM | Chat Model |
|---|---|---|
| **Input** | Text / String | List of Messages |
| **Output** | Text / String | AI Message |
| **Conversation structure** | Manual | Native message structure |
| **System instructions** | Usually embedded in prompt text | Separate system message |
| **Multi-turn chat** | Can be implemented manually | Natural fit |
| **Tool calling** | Less natural | Strong fit |

**A useful memory trick:**

```text
LLM
→ Text in
→ Text out

Chat Model
→ Messages in
→ Message out
```

> **Whatever we can do with an LLM can often also be done with a Chat Model, while the message-based abstraction is usually more convenient for modern conversational applications.**

# 🌍 **Chat Model Example — Translation**

### **Application**

```text
User
 ↓
English Sentence
 ↓
Chat Model
 ↓
Hindi Translation
```

### **Code**

````python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

model = ChatOpenAI()

messages = [
    SystemMessage(
        content="You are a helpful assistant that translates English to Hindi."
    ),
    HumanMessage(
        content="What is the capital of France?"
    )
]

response = model.invoke(messages)

print(response.content)
````

# 🧩 **Prompt**

> **Prompt = The input instructions, context, examples, constraints or task information supplied to the model.**

**A prompt can contain:**

```text
Role / Behavior
      +

      +

      +
```

# 📝 **Prompt Templates**

> **PromptTemplate = A reusable template that creates prompts dynamically from runtime variables.**

**Without a template:**

```python
llm.invoke("What is the capital of USA?")
```

For another country, we manually change the string.

**With a template:**

```text
Template:
"What is the capital of {country}?"

Runtime value:
country = "USA"

              ↓

Final Prompt:
"What is the capital of USA?"
```

---

# 🔄 **PromptTemplate Mental Model**

```text
Prompt Template
      +
Runtime Variables
      ↓
Final Prompt
```

**Formally:**

$$
\text{Final Prompt}
=
f(\text{Template},\text{Runtime Variables})
$$

---

# 🧱 **PromptTemplate Example**

````python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="What is the capital of {country}?"
)

prompt = prompt_template.invoke(
    {"country": "USA"}
)

print(prompt)
````

**The variable can be changed without changing the template:**

````python
prompt_template.invoke({"country": "India"})
prompt_template.invoke({"country": "France"})
prompt_template.invoke({"country": "Japan"})
````

---

# 🧠 **Multiple Variables**

**A template can contain multiple placeholders:**

````python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="What is the capital of {country} in {language}?"
)

prompt = prompt_template.invoke({
    "country": "USA",
    "language": "English"
})

print(prompt)
````

The values are supplied at runtime.

---

# 📌 **Why use PromptTemplate?**

**Without a prompt template:**

````python
user_input = "USA"

llm.invoke(
    f"What is the capital of {user_input}?"
)
````

This works, but as the application grows, prompts become harder to maintain.

**With a template:**

```text
Reusable Prompt
        ↓
Runtime Variables
        ↓
Generated Prompt
```

**Advantages:**

```text
Reusability
Maintainability
Consistency
Dynamic Input
Cleaner Application Logic
```

# 🧩 **Why do we need Chains?**

**Suppose we write the workflow manually:**

```python
prompt = prompt_template.invoke(
    {"country": "USA"}
)

```

There are multiple independent steps.

**LangChain lets us compose them:**

```python
chain = prompt_template | llm
```

**Now the workflow is reusable.**

```python
chain.invoke(
    {"country": "USA"}
)
```

# 🔄 **Chain Example**

````python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="What is the capital of {country}?"
)

prompt = prompt_template.invoke(
    {"country": "USA"}
)

print(prompt)
````

**Now invoke the model:**

````python
response = llm.invoke(prompt)

print(response)
````

**Instead of managing both steps separately, compose them:**

````python
chain = prompt_template | llm
````

**Then:**

````python
response = chain.invoke(
    {"country": "USA"}
)

print(response)
````

# 🧬 **LCEL — LangChain Expression Language**

> **LCEL = LangChain Expression Language**

LCEL provides a declarative way to compose LangChain components.

**The most recognizable operator is:**

```text
|
```

called the **pipe operator**.

---

# 🚰 **Pipe Operator (`|`)**

The pipe operator connects components.

```python
chain = comp1 | comp2 | comp3
```

The output of one component becomes the input of the next component.

```text
comp1
  ↓
comp2
  ↓
comp3
```

---

# 🔢 **Execution Order**

**The execution order is:**

> **Left → Right**

**For:**

```python
chain = comp1 | comp2 | comp3
```

**the conceptual execution is:**

```python
out1 = comp1.invoke(input)

out2 = comp2.invoke(out1)

out3 = comp3.invoke(out2)

final_output = out3
```

**Therefore:**

$$
\text{Final Output}
=
\text{comp3}(
\text{comp2}(
\text{comp1}(\text{Input})
))
$$

---

# 🔗 **Simple LCEL Example**

```python
chain = prompt_template | llm
```

**Then:**

```python
chain.invoke(
    {"country": "USA"}
)
```

**Conceptually:**

```text
{"country": "USA"}
        ↓
PromptTemplate
        ↓
"What is the capital of USA?"
        ↓
LLM
        ↓
AI / Text Response
```

# 🏃 **LangChain Runnables**

Many LangChain components implement a common runnable interface.

> **Runnable = A component that can be executed with standard methods such as `invoke`, `batch`, and `stream`.**

This gives different components a consistent execution interface.

**Examples include:**

```text
Prompt Template
Model
Output Parser
RunnableLambda
RunnablePassthrough
```

**A simplified mental model is:**

```text
Any Runnable
    │
    ├── invoke()
    ├── batch()
    └── stream()
```

---

# ▶️ **Runnable Execution Methods**

**The three important methods are:**

```text
.invoke()
.batch()
.stream()
```

---

# 1️⃣ **invoke()**

> **invoke() = Run a component once with a single input and return its output.**

**Example:**

```python
chain.invoke(
    {"country": "USA"}
)
```

**Conceptually:**

```text
Single Input
    ↓
Runnable
    ↓
Single Output
```

---

# 2️⃣ **batch()**

> **batch() = Run a runnable on multiple inputs and return multiple outputs.**

**Example:**

````python
chain.batch([
    {"country": "USA"},
    {"country": "India"},
    {"country": "France"}
])
````

**Conceptually:**

```text
Input 1 ─┐
Input 2 ─┼→ Runnable → Output 1
Input 3 ─┘            → Output 2
                       → Output 3
```

This is useful when processing many independent inputs.

---

# 3️⃣ **stream()**

> **stream() = Run a runnable and receive output incrementally as it becomes available.**

**For a model:**

````python
llm.stream(
    "What is the capital of France?"
)
````

**A useful intuition is:**

```text
Normal invoke()

Request
   ↓
[Wait...]
   ↓
Full Response


Streaming

Request
   ↓
Token
Token
Token
Token
...
```

Streaming improves perceived responsiveness because the application can start displaying output before the entire response has been produced.

---

# 🧪 **Streaming a Model**

````python
for token in llm.stream(
    "What is the capital of France?"
):
    print(token, end="")
````

Depending on the model abstraction, the streamed objects may contain content or other message information.

For a simple text completion model, the chunks can often be printed directly.

---

# 🧪 **Streaming a Chain**

````python
chain = prompt_template | llm

for chunk in chain.stream(
    {"country": "USA"}
):
    print(chunk, end="")
````

**Conceptually:**

```text
Input
 ↓
Prompt Template
 ↓
Model
 ↓
Chunk 1
 ↓
Chunk 2
 ↓
Chunk 3
 ↓
...
```

---

# 🆚 **invoke vs batch vs stream**

| Method | Input | Output | Main Use |
|---|---|---|---|
| **invoke()** | One input | One output | Normal execution |
| **batch()** | Multiple inputs | Multiple outputs | Bulk processing |
| **stream()** | One input | Incremental output | Real-time display |

Memory trick:

```text
invoke
→ One → One

batch
→ Many → Many

stream
→ One → Chunks
```

# 🔧 **RunnableLambda**

> **RunnableLambda = A runnable created from a normal Python function.**

**Example:**

````python
from langchain_core.runnables import RunnableLambda

def uppercase(text):
    return text.upper()

runnable = RunnableLambda(uppercase)

result = runnable.invoke("hello")

print(result)
````

**Output:**

```text
HELLO
```

**Mental model:**

```text
Input
 ↓
Python Function
 ↓
Output
```

---

# 🔗 **RunnableLambda inside a Chain**

````python
from langchain_core.runnables import RunnableLambda

def add_exclamation(text):
    return text + "!"

chain = RunnableLambda(add_exclamation)

print(chain.invoke("Hello"))
````

**Output:**

```text
Hello!
```

**Now combine it with other runnables:**

````python
chain = (
    RunnableLambda(lambda x: x.upper())
    | RunnableLambda(lambda x: x + "!")
)

print(chain.invoke("hello"))
````

**Conceptually:**

```text
"hello"
   ↓
.upper()
   ↓
"HELLO"
   ↓
+ "!"
   ↓
"HELLO!"
```

The power of LCEL comes from treating many application components as composable runnables.

```text

```

**or:**


This lets you build larger workflows from small reusable pieces.

# 🧠 **Important Rule of LCEL**

**The basic rule is:**

> **The output type of one component must be compatible with the input expected by the next component.**

**Conceptually:**

$$
\text{Output}_{n}
\rightarrow
\text{Input}_{n+1}
$$

**Therefore:**

```python
comp1 | comp2
```

**means:**

```text
comp1.output
      ↓
comp2.input
```

# 🤖 **Agents**

> **Agent = A model-driven system that can decide which tools or actions to use to accomplish a task.**

```text

```

Modern LangChain provides a `create_agent` abstraction for building agent applications.

**A current-style example is:**

````python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model="openai:gpt-5.5",
    tools=[get_weather],
    system_prompt="You are a helpful assistant."
)

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What's the weather in San Francisco?"
        }
    ]
})

print(result["messages"][-1].content_blocks)
````

# 🛠️ **Notebook Practice — Complete Mini Example**

**The following example combines:**

```text
PromptTemplate
      +
LLM
      +
LCEL
      +
invoke
      +
batch
      +
stream
```

### **1. Create the prompt**

````python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template="What is the capital of {country}?"
)
````

### **2. Create the model**

````python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-5-nano"
)
````

### **3. Build the chain**

````python
chain = prompt | model
````

### **4. Invoke once**

````python
response = chain.invoke({
    "country": "India"
})

print(response.content)
````

### **5. Batch multiple requests**

````python
responses = chain.batch([
    {"country": "India"},
    {"country": "France"},
    {"country": "Japan"}
])

for response in responses:
    print(response.content)
````

### **6. Stream the response**

````python
for chunk in chain.stream({
    "country": "India"
}):
    print(chunk.content, end="")
````

# 🧪 **Notebook Practice — Simple Runnable Chain**

````python
from langchain_core.runnables import RunnableLambda

chain = (
    RunnableLambda(lambda x: x.strip())
    | RunnableLambda(lambda x: x.lower())
    | RunnableLambda(lambda x: f"Processed: {x}")
)

result = chain.invoke(
    "   HELLO LANGCHAIN   "
)

print(result)
````

**Output:**

# ⚠️ **Common Beginner Mistakes**

## **1. Thinking LangChain is an LLM**

```text
LangChain
≠
LLM
```

LangChain is a framework / ecosystem for building applications around models.

---

## **2. Using a Chat Model Like a Plain Text Completion API**

A chat model is designed around messages.

```text
System
Human
AI
Tool
```

Understanding messages makes modern LangChain applications much easier.

---

## **3. Thinking PromptTemplate is the Prompt Itself**

```text
PromptTemplate
→ Template

invoke(...)
→ Concrete Prompt / Messages
```

**Example:**

```text
"What is the capital of {country}?"
```

is the template.

```text
"What is the capital of India?"
```

is the generated prompt.

---

## **4. Forgetting the Input Type**

**For:**

```python
prompt_template = PromptTemplate(
    template="What is the capital of {country}?"
)
```

**the runnable expects:**

```python
{
    "country": "India"
}
```

**not:**

```python
"India"
```

unless the component has been designed accordingly.

---

## **5. Thinking `|` is Just Python OR**

**In LCEL:**

```python
chain = prompt | model
```

the pipe represents component composition.

```text
Prompt Output
      ↓
Model Input
```

It is not logical OR.

---

## **6. Ignoring Output Types**

A chat model usually returns an AI message object.

**So:**

```python
response = model.invoke(...)
```

**and then:**

```python
response.content
```

may be needed when you want the textual content.

---

## **7. Confusing `invoke`, `batch`, and `stream`**

```text
invoke
→ One input / one result

batch
→ Many inputs / many results

stream
→ Incremental output
```

---

## **8. Copying Old LangChain Tutorials Blindly**

LangChain APIs evolve quickly.

**Older tutorials may use:**

```text
LLMChain
RetrievalQA
Old agent constructors
Older package names
```

**For new projects:**

> **Prefer the current official LangChain documentation and current package APIs.**

> **4. Prompt templates separate reusable prompt structure from runtime variables.**

> **5. `ChatPromptTemplate` is designed for structured messages such as system and human messages.**

> **6. A chain connects multiple application components into a reusable workflow.**

> **7. LCEL uses the pipe operator `|` to compose components from left to right.**

> **8. A runnable provides a common execution interface such as `invoke()`, `batch()`, and `stream()`.**

> **9. `RunnableLambda` wraps normal Python functions so they can participate in runnable pipelines.**

> **10. `RunnablePassthrough` forwards its input without changing it.**

> **11. RAG combines retrieval with generation so external information can be supplied to a model at runtime.**

> **12. Tools allow an application to expose external capabilities such as APIs, databases and Python functions.**

> **13. Agents use models to decide which actions or tools to take.**

> **14. LangGraph is useful when agentic workflows require explicit state, routing and control.**

> **15. LangSmith provides tracing, debugging, observability and evaluation capabilities.**

> **16. Direct provider SDKs are often enough for simple applications; LangChain becomes more useful as component composition grows.**

> **17. Legacy chain classes still appear in older tutorials and codebases, but modern applications should generally follow current LangChain composition patterns.**
