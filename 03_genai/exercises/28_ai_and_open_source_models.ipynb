# 🧠 **What is an LLM?**

> **LLM (Large Language Model) = A neural network trained on a large amount of data to understand and generate language by predicting token sequences.**

**An LLM can be used for tasks such as:**

```text
Text Generation
Question Answering
Summarization
Translation
Classification
Code Generation
Reasoning
Tool Use
Structured Output
Multi-Modal Tasks
```

**A useful mental model is:**

```text
User Input
    ↓
  Tokens
    ↓
    LLM
    ↓
Predicted Tokens
    ↓
  Response
```

---

# ⚙️ **Factors to Consider While Selecting an LLM**

Choosing an LLM should not be based only on benchmark scores. The best model depends on the application's requirements.

```text
                    LLM Selection
                         │
       ┌─────────────────┼─────────────────┐
       ↓                 ↓                 ↓
 Capability            Cost             Security
       │                 │                 │
       ├── Tool Use     ├── Input         ├── Data Privacy
       ├── Reasoning    └── Output        ├── Compliance
       ├── Structured                      └── Deployment Control
       │   Output
       └── Multi-Modal

       ┌─────────────────┼─────────────────┐
       ↓                 ↓                 ↓
 Context Window       Latency          Performance
                                            │
                                    ┌───────┼────────┐
                                    ↓       ↓        ↓
                                 Accuracy Reliability Consistency
```

## 🛠️ **1. Capability**

Check what the model can actually do for the target application.

```text
Tool Use
→ Can the model call tools / functions?

Reasoning
→ Can it handle multi-step reasoning tasks?

Structured Output
→ Can it reliably return JSON / schema-constrained output?

Multi-Modal
→ Can it work with text, images, audio, or other modalities?
```

> **Choose capability according to the task. A simpler model can be a better choice when advanced capabilities are unnecessary.**

---

## 💰 **2. Cost**

For API-based models, cost commonly depends on token usage and the provider's pricing model.

```text
More Input Tokens
        ↓
Higher Input Cost

More Output Tokens
        ↓
Higher Output Cost
```

**Also consider:**

```text
Caching
Batch Processing
Number of Requests
Model Size
Infrastructure Cost
```

> **The cheapest model is not always the cheapest system. A model that needs fewer retries or produces shorter answers can reduce total application cost.**

---

## 🔐 **3. Security**

Security becomes important when the application handles sensitive or proprietary information.

**Consider:**

```text
Data Privacy
Access Control
Encryption
Compliance Requirements
Data Residency
Retention Policies
Self-Hosting Requirements
```

> **Hosted API vs self-hosted model is partly a security and control decision, not only a cost decision.**

---

## 📚 **4. Context Window**

> **Context Window = The amount of tokenized information a model can process in a single request according to its configured limits.**

**It can include:**

```text
System Instructions
Conversation History
User Prompt
Retrieved Documents
Tool Results
Other Input Content
```

A larger context window is useful for long documents and long conversations, but a larger limit does not automatically mean every application should send more context.

---

## ⚡ **5. Latency**

> **Latency = How long the user waits for the model response.**

**Latency matters in:**

```text
Chat Applications
Voice Applications
Real-Time Agents
Customer Support
Interactive Applications
```

For these systems, throughput and time-to-first-token can also matter in addition to total response time.

---

## 📊 **6. Performance**

Performance should be evaluated against the actual task.

```text
Accuracy
→ Does the model produce correct answers?

Reliability
→ Does it work correctly across many cases?

Consistency
→ Does it produce stable results for similar inputs?
```

**Other useful evaluation dimensions include:**

```text
Instruction Following
Tool-Calling Accuracy
Structured Output Reliability
Hallucination Rate
Domain Performance
Safety
```

> **Benchmark performance is a useful signal, but task-specific evaluation is more important for production decisions.**

# 💰 **LLM Costing**

For token-priced APIs, billing commonly distinguishes between **input tokens** and **output tokens**.

```text
                 API Request
                     │
          ┌──────────┴──────────┐
          ↓                     ↓
      Input Tokens         Output Tokens
          ↓                     ↓
       Prompt              Model Response
          ↓                     ↓
       Input Cost          Output Cost
```

## 📥 **Input Tokens**

> **Input tokens = Tokens sent to the model as part of the request.**

**They can include:**

```text
System Prompt
User Prompt
Conversation History
Retrieved RAG Context
Tool Results
```

## 📤 **Output Tokens**

> **Output tokens = Tokens generated by the model in its response.**

**Therefore, a simplified cost model is:**

$$
\text{Total Cost}
\approx
\text{Input Tokens}\times\text{Input Price}
+
\text{Output Tokens}\times\text{Output Price}
$$

The exact billing model depends on the provider and model.

---

# 🧊 **Cached Input**

> **Cached input = Reused input content that a provider can process through prompt caching under its caching rules.**

When eligible input is served from cache, providers may charge a lower cached-input rate.

**Conceptually:**

```text
First Request
    ↓
Prompt / Input
    ↓
Processed & Cached

Later Request
    ↓
Repeated Prefix / Input
    ↓
Cache Hit
    ↓
Lower Cached-Input Price
```

> **Caching is provider-specific. Eligibility, cache duration, minimum token requirements, and discounts depend on the provider and model.**

---

# 📦 **Batch API**

> **Batch API = An asynchronous API workflow where many requests are submitted together and processed within a defined processing window.**

For OpenAI, the Batch API currently offers a **50% discount** compared with synchronous API pricing and uses a processing window of up to 24 hours for the documented Batch API configuration.

### **Conceptual Flow**

```text
Many Requests
      ↓
Create Batch
      ↓
Asynchronous Processing
      ↓
Wait for Completion
      ↓
Retrieve Results
```

### ✅ **When Batch Processing is Useful**

```text
Offline Evaluation
Large-Scale Classification
Embedding Jobs
Dataset Processing
Nightly / Scheduled Workloads
```

### ⚠️ **Not Ideal For**

```text
Real-Time Chat
Interactive Applications
Low-Latency User Requests
```

> **Memory trick:**

```text
Normal API
→ Fast / Interactive

Batch API
→ Cheap / Asynchronous
```

# 💾 **How to Calculate Required Model Storage?**

Before loading a model, it is useful to estimate how much memory its parameters require.

## 🔢 **Bits and Bytes**

```text
1 bit
→ Can represent 0 or 1

8 bits
→ 1 byte
```

**Examples:**

```text
10
→ 2 bits

11
→ 2 bits

111
→ 3 bits

1000001
→ 7 bits
```

The important idea is that **the number of bits determines how many distinct binary states can be represented**.

> ⚠️ **Correction:** A bit is a unit of information. It is not accurate to say that “1 bit = 1 transistor” as a general rule. Real memory and storage hardware use more complex physical structures.

---

# 📏 **Storage Units**

**For decimal storage calculations:**

```text
1 KB → 10³ bytes
1 MB → 10⁶ bytes
1 GB → 10⁹ bytes
1 TB → 10¹² bytes
```

> **Note:** Operating systems and hardware documentation may also use binary units such as KiB, MiB, GiB, and TiB, where the base is 1024.

---

# 🔢 **Data Types and Memory**

The storage required for one parameter depends heavily on the numerical precision used to store it.

| Data Type | Bits / Parameter | Approx. Bytes / Parameter |
|---|---:|---:|
| **int8** | 8 bits | 1 byte |
| **float16** | 16 bits | 2 bytes |
| **float32** | 32 bits | 4 bytes |
| **float64** | 64 bits | 8 bytes |
| **4-bit quantized** | 4 bits | 0.5 bytes |

### 🧮 **Basic Formula**

**For a rough raw-weight estimate:**

$$
\text{Memory} \approx
\text{Number of Parameters}
\times
\text{Bytes per Parameter}
$$

> **Important:** This is only a rough estimate. Real model memory is higher because of metadata, buffers, temporary activations, KV cache, non-quantized layers, and framework/runtime overhead.

---

# 🧮 **Example: 8 Billion Parameter Model**

**Suppose a model has:**

```text
8 Billion Parameters
```

### **float32**

```text
4 bytes / parameter

8 × 10⁹ × 4
= 32 × 10⁹ bytes
≈ 32 GB
```

### **float16**

```text
2 bytes / parameter

8 × 10⁹ × 2
= 16 × 10⁹ bytes
≈ 16 GB
```

### **int8**

```text
1 byte / parameter

8 × 10⁹ × 1
= 8 × 10⁹ bytes
≈ 8 GB
```

### **4-bit Quantization**

```text
0.5 bytes / parameter

8 × 10⁹ × 0.5
= 4 × 10⁹ bytes
≈ 4 GB
```

**Conceptually:**

```text
8B Model
   │
   ├── float32 → ~32 GB
   ├── float16 → ~16 GB
   ├── int8    → ~8 GB
   └── 4-bit   → ~4 GB
```

> **Memory trick:**

```text
32-bit → 4 bytes
16-bit → 2 bytes
 8-bit → 1 byte
 4-bit → 0.5 byte
```

# 🧰 **Load a Llama Model with Transformers**

````python
from transformers import AutoTokenizer, AutoModelForCausalLM
````

# 🗜️ **Quantization**

> **Quantization = Representing model values with lower precision so that model memory usage and computational requirements can be reduced.**

**Conceptually:**

```text
Higher Precision
      ↓
More Memory
      ↓
More Hardware Required

Lower Precision
      ↓
Less Memory
      ↓
Easier Deployment
```

Hugging Face Transformers supports multiple quantization approaches, including 8-bit and 4-bit loading through **`bitsandbytes`**.

---

# 🔢 **Precision Comparison**

**For a simplified 8B-parameter raw-weight estimate:**

```text
float32
→ 32 bits / parameter
→ ~32 GB

float16
→ 16 bits / parameter
→ ~16 GB

int8
→ 8 bits / parameter
→ ~8 GB

4-bit quantization
→ 4 bits / parameter
→ ~4 GB
```

> **Important: `int4` / 4-bit** quantization is not the same as saying that an IEEE-standard “float4” type is being used. Modern quantization methods can use formats such as FP4 or NF4 depending on the method and library.

Hugging Face's current **`bitsandbytes`** documentation supports **4-bit quantization** and provides configurations such as NF4 for 4-bit workflows.

---

# 📚 **BitsAndBytes**

**`bitsandbytes`** provides convenient quantization support for loading models in 8-bit or 4-bit form with Transformers.

### **Install**

```python
!pip install -U bitsandbytes
```

### **Import**

````python
from transformers import BitsAndBytesConfig
````

---

# 🧪 **Create a 4-Bit Quantization Configuration**

````python
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
)
````

### **Load the Quantized Model**

````python
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    token=hf_token,
    quantization_config=bnb_config,
    device_map="auto"
)
````

This tells Transformers to load the supported model using 4-bit quantization.

> **Why use `device_map="auto"`?** It allows Transformers / Accelerate to place the model across available compute devices according to the loading configuration. Hugging Face's quantization documentation commonly uses this pattern for large-model loading.

---

# 💾 **Save the Quantized Model**

````python
model.save_pretrained("llama-4bit")
tokenizer.save_pretrained("llama-4bit")
````

Now the quantized model can be stored in a local directory and reused without repeating the original model-download step, subject to the usual model/configuration requirements.

# 🏭 **Create a Text Generation Pipeline**

Transformers provides a high-level **`pipeline()`** interface for inference.

````python
from transformers import pipeline
````

### **Create the Pipeline**

````python
pipe = pipeline(
    task="text-generation",
    model=model,
    tokenizer=tokenizer,
    return_full_text=False
)
````

### **Run Inference**

````python
response = pipe(
    "What is the capital of India?"
)

print(response)
````

The Hugging Face Llama model card also documents using Transformers' **`pipeline`** abstraction for text-generation inference.

For open-weight models, I can either run them locally using Transformers or use a hosted inference provider. If the model is too large for my available hardware, I can use quantization, such as 8-bit or 4-bit loading, to reduce the memory requirement while accepting the potential quality trade-off.”**
