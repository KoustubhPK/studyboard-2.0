# 🏛️ **What is an LLM?**

# **LLM = Large Language Model**

An LLM is a machine-learning model trained on large amounts of language data to model the probability of token sequences.

**For an autoregressive language model, the central idea is:**

$$
P(x_1,x_2,\ldots,x_T)
=
\prod_{t=1}^{T}
P(x_t \mid x_{<t})
$$

**where:**

$$
x_{<t} = (x_1,x_2,\ldots,x_{t-1})
$$

**So, at each position, the model estimates:**

> **"Given everything I have seen so far, what token is most likely to come next?"**

---

# 🧩 **Why is it called "Large"?**

**The word **Large** usually refers to scale in several dimensions:**

```text
Large
 ├── Number of parameters
 ├── Training data
 ├── Compute used for training
 └── Model / context capability
```

A model with many parameters has a large number of learned numerical values.

These parameters are adjusted during training so that the model becomes better at predicting tokens.

---

# 🔢 **Parameters**

> **Parameters = Learnable numerical values inside a model.**

A neural network learns by updating parameters.

**Conceptually:**

```text
Training Data
      ↓
Model
      ↓
Prediction
      ↓
Loss
      ↓
Gradient
      ↓
Update Parameters
      ↓
Better Model
```

The number of parameters is often used as a rough indicator of model scale, but:

> **More parameters does not automatically mean a better model.**

Architecture, data quality, training method, inference strategy, and model optimization also matter.

**A token can be:**

```text
A whole word

Punctuation

A special token
```

**For example:**

```text

```

may be represented by multiple subword tokens depending on the tokenizer.

# 🧠 **Embeddings**

An embedding converts text into a numerical vector representation.

**Conceptually:**

```text
Text
 ↓
Embedding Model
 ↓
Vector
```

**For example:**

```text
"Apple is a fruit"
        ↓

```

The vector contains information useful for measuring semantic relationships.

---

# 📐 **Semantic Similarity**

Two pieces of text with similar meaning tend to have embeddings that are close according to a chosen similarity measure.

**A commonly used similarity measure is cosine similarity:**

$$
\operatorname{cosine\ similarity}(a,b)
=
\frac{a\cdot b}
{\|a\|\|b\|}
$$

**This idea is widely used in:**

```text
Semantic Search
RAG
Document Retrieval
Recommendation
Clustering
Deduplication
```

> **Embedding models and generative LLMs are related, but they are not the same type of model.**

# 🧠 **Self-Attention**

Self-attention allows each token representation to incorporate information from other relevant tokens.

**The standard formulation is:**

$$
\operatorname{Attention}(Q,K,V)
=
\operatorname{softmax}
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
$$

**where:**

```text
Q → Query
K → Key
V → Value
d_k → Key dimension
```

# 🔁 **Autoregressive Generation**

A decoder-style LLM usually generates text one token at a time.

**Conceptually:**

```text
Prompt
  ↓
Predict Token 1
  ↓
Context + Token 1
  ↓
Predict Token 2
  ↓
Context + Token 1 + Token 2
  ↓
...
```

**For example:**

```text
Input:
"What is the capital of"

Prediction:
"India"

Next context:
"What is the capital of India"

Prediction:
"?"

...
```

---

# 🎲 **Next-Token Probabilities**

At every generation step, the model produces a probability distribution over the vocabulary.

**Conceptually:**

```text
Prompt
  ↓
Model
  ↓
Logits
  ↓
Softmax
  ↓
Probability for each token
```

**The probability distribution can be written as:**

$$
P(x_t \mid x_{<t})
$$

The decoding system then chooses the next token according to the configured generation strategy.

---

# 🎛️ **Generation Parameters**

**Common generation controls include:**

```text
Temperature
Top-p
Maximum output tokens
Stop sequences
Seed / deterministic controls (where supported)
```

### **Temperature**

Temperature changes the sharpness of the probability distribution.

**A simplified expression is:**

$$
P_i
=
\frac{\exp(z_i/T)}
{\sum_j \exp(z_j/T)}
$$

**where:**

```text
z_i → Logit
T   → Temperature
```

**Intuition:**

```text
Lower temperature
→ More concentrated distribution
→ More deterministic behavior

Higher temperature
→ Flatter distribution
→ More variation
```

> Temperature behavior and parameter availability depend on the specific model/API.

---

# 🧮 **Logits and Softmax**

**The final model layer typically produces logits:**

$$
z_1,z_2,\ldots,z_V
$$

where \(V\) is the vocabulary size.

**Softmax converts these logits into probabilities:**

$$
P(x_i)
=
\frac{e^{z_i}}
{\sum_{j=1}^{V}e^{z_j}}
$$

**The probabilities sum to:**

$$
\sum_{i=1}^{V}P(x_i)=1
$$

# 📚 **How LLMs Are Trained**

**A useful simplified training pipeline is:**

```text

      ↓

      ↓

      ↓
Base / Foundation Model
      ↓

      ↓

      ↓

```

The exact modern pipeline varies by organization and model family.

---

# 1️⃣ **Pre-training**

> **Pre-training = Learning broad language patterns, concepts, and statistical relationships from a very large corpus.**

For autoregressive models, a central objective is next-token prediction.

**Example:**

```text
"We are"
    ↓
"training"
```

**Then:**

```text
"We are training"
         ↓
"LLM"
```

**For a token sequence:**

$$
x_1,x_2,\ldots,x_T
$$

**the model learns:**

$$
P(x_t\mid x_{<t})
$$

**A common objective is negative log-likelihood / cross-entropy loss:**

$$
\mathcal{L}
=
-\sum_{t=1}^{T}
\log P(x_t\mid x_{<t})
$$

The model updates its parameters to reduce this loss.

---

# 📦 **What does the model learn during pre-training?**

**Pre-training can lead to capabilities such as:**

```text
Language patterns
Grammar
Syntax
Semantics
Facts / information patterns
Code patterns
Style patterns
Reasoning-related patterns
Domain-specific associations
```

**However:**

> **The model's behavior is determined by the training distribution and optimization process; it should not be interpreted as a perfect database of facts.**

---

# 🧱 **Base / Foundation Model**

> **Base Model / Foundation Model = A broadly pretrained model before task-specific or instruction-specific post-training.**

A base model may be very capable at continuing text but may not naturally behave like a helpful assistant.

**For example:**

```text
Prompt:
"Summarize this article."

Base model:
May continue or imitate text patterns.

Instruction-tuned model:
More likely to produce a direct summary.
```

---

# 2️⃣ **Supervised Fine-Tuning (SFT)**

# **SFT = Supervised Fine-Tuning**

**SFT trains a pretrained model on curated examples of:**

```text
Instruction / Prompt
        ↓
Desired Response
```

**Example:**

| Prompt | Target Response |
|---|---|
| **What is the capital of India?** | **New Delhi.** |
| **Define an LLM.** | **An LLM is a model trained to model language.** |
| **Write Python code.** | `print("Hello World")` |

---

# 🎯 **Why do we need SFT?**

**Pre-training primarily teaches:**

> **"What token is likely to come next?"**

**Instruction tuning pushes the model toward:**

> **"How should I respond to this instruction?"**

**So:**

```text
Pre-training
→ Broad language capability

SFT
→ Instruction-following behavior
```

---

# 3️⃣ **Preference / Alignment Training**

A post-training pipeline can use preference information to push model behavior toward outputs that are considered more useful, safe, or appropriate.

**A simplified pipeline is:**

```text
Prompt
   ↓
Candidate Responses
   ↓
Human / AI Preference Data
   ↓
Preference Learning
   ↓
Model Optimization
```

---

# 🏆 **RLHF**

# **RLHF = Reinforcement Learning From Human Feedback**

In a classical RLHF pipeline, humans compare or rank candidate responses.

**Example:**

| Prompt | Response | Preference |
|---|---|---:|
| **Prompt** | Response 1 | Medium |
| **Prompt** | Response 2 | High |
| **Prompt** | Response 3 | Low |

The preference data can be used to train a **reward model**.

**Conceptually:**

```text
Prompt + Response
       ↓
Reward Model
       ↓
Reward
```

A reinforcement-learning optimizer can then use that reward signal to update the language-model policy.

**A simplified view is:**

```text
Prompt
  ↓
LLM / Policy
  ↓
Response
  ↓
Reward Model
  ↓
Reward
  ↓
Policy Update
  ↓
Better Expected Reward
```

---

# ⚠️ **Important: RLHF is not the only alignment method**

**Modern LLM post-training can use several methods, including:**

```text
SFT
RLHF
Preference Optimization
DPO
RLAIF
Rejection Sampling
Distillation
Other reward / policy optimization methods
```

**Therefore:**

> **"LLM alignment" is broader than RLHF alone.**

# 💾 **KV Cache**

During autoregressive decoding, previously calculated Key and Value states can be stored and reused.

```text
Previous Tokens
      ↓
Previously calculated K/V
      ↓
   KV Cache
      ↓
Reuse during next generation step
```

**Without KV caching:**

```text
Repeatedly compute old K/V information
```

**With KV caching:**

```text
Reuse old K/V information
```

> **KV Cache trades memory for faster autoregressive inference.**

Your attention notes cover this in detail.

---

# 🧩 **MHA vs MQA vs GQA**

**The three common attention designs can be summarized as:**

```text
MHA
→ Each attention head has its own K/V

GQA
→ Groups of query heads share K/V

MQA
→ Query heads share a common K/V
```

**Conceptually:**

```text
MHA
Q1 → K1,V1
Q2 → K2,V2
Q3 → K3,V3

GQA
Q1 ─┐
Q2 ─┤ → K/V Group 1
Q3 ─┘

MQA
Q1 ─┐
Q2 ─┤
Q3 ─┤ → Shared K/V
Q4 ─┘
```

**The trade-off is roughly:**

$$
\text{K/V Sharing} \uparrow
\quad\Longrightarrow\quad
\text{KV Cache Memory} \downarrow
$$

while the quality trade-off depends on the model architecture and training setup.

---

# 🪟 **Sliding Window Attention**

Sliding-window attention limits the tokens that each position directly attends to.

```text
Full / Dense Attention
→ Attend to many or all allowed previous positions

Sliding Window Attention
→ Attend to a local range
```

**Conceptually:**

```text
Token 1 → 1
Token 2 → 1,2
Token 3 → 1,2,3
Token 4 → 2,3,4
Token 5 → 3,4,5
...
```

> **Local direct attention can reduce computation and memory pressure for long sequences.**

---

# ⚡ **Flash Attention**

Flash Attention is best understood as an **IO-aware implementation of attention**.

It does not replace the underlying attention idea with a different mathematical objective.

**Conceptually:**

```text
Standard implementation
→ More intermediate materialization
→ More memory traffic

Flash Attention
→ Tiled / blocked computation
→ Less unnecessary memory traffic
→ Better memory efficiency
```

**Therefore:**

```text
MHA / MQA / GQA
→ Change K/V sharing

Sliding Window Attention
→ Change attention connectivity

Flash Attention
→ Change how attention is implemented efficiently
```

> **Flash Attention and a model family named "Flash" are different concepts.**

|---|---|---|

| **Internet** | Typically required to call hosted API | Can run offline after the model and dependencies are available |
| **Operational Burden** | Lower | Higher |
| **Control** | Lower | Higher |

# 🧪 **Reusable Helper Function**

**A simple notebook helper can be:**

````python
from openai import OpenAI

client = OpenAI()

def ask_llm(prompt: str, model: str = "gpt-5") -> str:
    response = client.responses.create(
        model=model,
        input=prompt
    )
    return response.output_text

print(ask_llm("Explain transformers in simple language."))
````

# 🎯 **Prompt Engineering**

# 📚 **RAG — Retrieval-Augmented Generation**

A model's pretrained knowledge has limitations.

**For example, an application may need:**


Instead of trying to put all information into the model's weights, a RAG system can retrieve the relevant information at request time.

# 🧮 **RAG Similarity**

**If embeddings are used with cosine similarity:**

$$
\operatorname{sim}(q,d)
=
\frac{q\cdot d}
{\|q\|\|d\|}
$$

**where:**

```text
q → Query embedding
d → Document embedding
```

The retriever can select documents with high similarity.

# 📚 **Key Takeaways**

> **1. An LLM models language as a probability distribution over token sequences.**

> **2. Tokens are the basic units processed by the model, and token count directly affects context, latency, and often cost.**

> **3. Transformers use self-attention to let token representations interact with one another.**

> **4. Pre-training builds broad language capability; SFT teaches instruction following; alignment/post-training shapes preferred behavior.**

> **5. Inference uses learned parameters to generate outputs without updating model weights.**

> **6. KV Cache reduces repeated computation during autoregressive decoding at the cost of additional memory.**

> **7. MHA, GQA, and MQA mainly differ in how Key and Value representations are shared across attention heads.**

> **8. Flash Attention is an efficient implementation of attention, while sliding-window attention changes which tokens directly participate in attention.**

> **9. Commercial / provider-hosted models reduce infrastructure burden, while open-weight models provide greater control at the cost of deployment responsibility.**

> **10. Open-weight does not automatically mean completely free, completely private, or fully open-source.**

> **11. API keys are secrets and should be stored outside source code.**

> **12. LangChain is an application framework / ecosystem, not an LLM itself.**

> **13. RAG supplies external context to the model at runtime instead of requiring all information to be stored in the model weights.**

> **14. Tools extend what an LLM application can do; agents can decide when and how to use those tools.**

> **15. Production LLM systems require evaluation, observability, security, cost control, and reliability — not only a model API.**
