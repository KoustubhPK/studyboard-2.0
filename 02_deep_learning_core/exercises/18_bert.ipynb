# 🤖 **BERT — Bidirectional Encoder Representations from Transformers**

> **BERT = Bidirectional Encoder Representations from Transformers**

BERT is a **Transformer-based, encoder-only language representation model** designed to learn rich contextual representations of text.

**Its key idea is:**

> **Read the context from both the left and right directions simultaneously.**

# 🧠 **Language Model**

## **Language Model**

A **Language Model** estimates the probability of a token occurring in a sequence.

> **Language Model → learns patterns and structure of language.**

**Conceptually:**

```text
Previous Context
      ↓
Language Model
      ↓
Probability of Next / Missing Token
```

---

# 🔄 **Transfer Learning in NLP**

The major idea behind modern NLP is to first learn general language representations and then reuse those representations for downstream tasks.

```text
Large General Dataset
        ↓
Pretraining
        ↓
Pretrained Language Model
        ↓
Fine-Tuning
        ↓
Task-Specific Model
```

This is the same general **pretrain → fine-tune** idea used in modern Transformer workflows. Your Hugging Face notes describe fine-tuning as continuing training from pretrained weights on a task-specific dataset.

# 🚀 **2018 — OpenAI GPT**

## **GPT — Generative Pretrained Transformer**

GPT is based on the **Decoder** side of the Transformer architecture.

```text
GPT
 ↓
Transformer Decoder
 ↓
Autoregressive / Left-to-Right Processing
```

Because GPT uses **masked self-attention**, each token attends only to the appropriate previous context during autoregressive language modeling.

**Example:**

```text
Apple is a great company
```

A left-to-right language model can process the sequence like:

```text
Apple
  ↓
Apple is
  ↓
Apple is a
  ↓
Apple is a great
  ↓
Apple is a great company
```

# 🔵 **2018 — Google BERT**

## **BERT**

**BERT = Bidirectional Encoder Representations from Transformers**

BERT was released by Google AI in **2018**.

BERT uses the **Encoder** part of the Transformer architecture.

```text
BERT
 ↓
Transformer Encoder
 ↓
Bidirectional Contextual Representation
```

### **What does Bidirectional mean?**

A BERT token representation can incorporate information from both sides of the sentence.

```text
Left Context  ←  Target Token  →  Right Context
```

**For example:**

```text
The bank is near the river
```

The representation of **bank** can use context from both:

```text
The bank
   ↑
Left context

bank is near the river
      ↑
Right context
```

| Main direction | Left-to-right / autoregressive | Bidirectional |
| Main pretraining idea | Next-token prediction | Masked Language Modeling + NSP in original BERT |
| Typical strength | Text generation | Language understanding / representations |
| Common use | Generation | Classification, NER, QA, etc. |

```text

 ↓

```

```text
BERT

 ↓

```

# 🧩 **BERT Architecture**

BERT is an **encoder-only Transformer architecture**.

```text
Input Text
    ↓
Tokenizer
    ↓
Input IDs + Token Type IDs + Attention Mask
    ↓
Token + Segment + Position Embeddings
    ↓
BERT Encoder Layers
    ↓
Contextual Representations
    ↓
Task-Specific Head
    ↓
Prediction
```

**A simplified view is:**

```text
                 BERT
                   │
        ┌──────────┴──────────┐
        │                     │
   Input Embeddings      Encoder Stack
                              │
                   ┌──────────┴──────────┐
                   │                     │
              Self-Attention       Feed Forward
                   │                     │
                   └──────────┬──────────┘
                              ↓
                    Contextual Vectors
```

---

# 🧱 **Inside a BERT Encoder Layer**

A Transformer encoder layer contains the main building blocks:

```text
Input
  ↓
Multi-Head Self-Attention
  ↓
Add & Layer Normalization
  ↓
Feed-Forward Network
  ↓
Add & Layer Normalization
  ↓
Output
```

The important idea is that **self-attention allows every token to build a representation using the other relevant tokens in the sequence**.

---

# 👀 **Self-Attention in BERT**

**For a sentence such as:**

```text
The animal didn't cross the street because it was tired.
```

BERT can use the context around a token such as **it** to produce a richer representation.

**Conceptually:**

```text
The animal didn't cross the street because it was tired.
  ↑     ↑         ↑                      ↑
  └─────┴─────────┴──── context ────────┘
```

The output for each token is therefore a **contextual representation**, rather than an isolated word representation.

---

# 🧠 **Context Vector**

The output corresponding to a token is a contextual vector.

```text
Input Tokens
     ↓
BERT Encoder
     ↓
V1   V2   V3   V4   ...
```

Each vector contains information influenced by the surrounding context.

**For a sequence:**

```text
[CLS] We are learning Transformer [SEP]
```

**we can think of the output as:**

```text
V1   V2   V3   V4   V5   V6
 ↓    ↓    ↓    ↓    ↓    ↓
[CLS] We  are learning Transformer [SEP]
```

---

# 🎭 **Masked Language Model (MLM)**

One of the most important ideas in BERT pretraining is **Masked Language Modeling**.

The model hides some tokens and learns to predict the missing token from the surrounding context.

### **Example**

```text
We are [MASK] Transformer.
```

**Input:**

```text
[We] [are] [MASK] [Transformer]
```

**The encoder processes the complete contextual input and produces vectors:**

```text
Input
 ↓
BERT Encoder
 ↓
V1  V2  V3  V4
       ↑
   [MASK] position
       ↓
Prediction
```

The objective is to predict the original token at the masked position.

```text
We are [MASK] Transformer.
           ↓
        learning
```

---

# 🎯 **Why MLM?**

Masked Language Modeling encourages the model to learn relationships between a token and **both its left and right context**.

```text
Left Context
     ↓
[  We  are  MASK  Transformer  ]
                    ↑
               Right Context
```

This is a key reason BERT is described as **bidirectional**.

# 🧪 **BERT Pretraining Tasks**

The original BERT training setup used two main pretraining objectives:

1. **Masked Language Model (MLM)**
2. **Next Sentence Prediction (NSP)**

```text
                 BERT Pretraining
                       │
             ┌─────────┴─────────┐
             ↓                   ↓
            MLM                 NSP
             ↓                   ↓
    Predict masked token   Predict sentence relation
```

---

# 📝 **1. Masked Language Model (MLM)**

### **Task**

> **Predict the missing / masked token.**

**Example:**

```text
The cat is sitting on the [MASK].
```

**Possible target:**

```text
mat
```

**Workflow:**

```text
Text
 ↓
Mask Some Tokens
 ↓
BERT Encoder
 ↓
Contextual Vectors
 ↓
Masked-Token Prediction
```

---

# 🔗 **2. Next Sentence Prediction (NSP)**

NSP is a **binary classification task** used in the original BERT pretraining setup.

The model receives two sentences and predicts whether the second sentence naturally follows the first.

### **Labels**

```text
isNextSentence
NotNextSentence
```

---

# 🧩 **BERT Input Format for NSP**

BERT uses special tokens to combine two sentences.

```text
[CLS] First sentence [SEP] Second sentence [SEP]
```

**Example:**

```text
[CLS] We are learning Transformer [SEP] BERT uses an Encoder [SEP]
```

---

# 🔵 **[CLS] Token**

`[CLS]` means **Classification Token**.

It is placed at the beginning of the sequence.

```text
[CLS] First sentence [SEP] Second sentence [SEP]
  ↑
Classification representation
```

For sentence-level tasks, the representation associated with `[CLS]` can be passed to a classification layer.

---

# 🟠 **[SEP] Token**

`[SEP]` means **Separator Token**.

It is used to separate sequences and mark the end of a sentence segment.

```text
[CLS] Sentence A [SEP] Sentence B [SEP]
                    ↑              ↑
                Separator       Final boundary
```

---

```text

                         ↓
                    BERT Encoder
                         ↓
                 Contextual Vectors
                         ↓
                 [CLS] Representation
                         ↓
                  Classification Head
                         ↓
          ┌──────────────┴──────────────┐
          ↓                             ↓
     isNextSentence              NotNextSentence
```

---

# 🏷️ **Binary Classification with BERT**

**For a binary classification task:**

```text

                           ↓
                         BERT
                           ↓

                           ↓
                     [CLS] Vector
```

# 🧱 **BERT Input Embeddings**

BERT represents an input token using information from multiple embedding components.

**Conceptually:**

```text
                 Token Embedding
                       +
                Segment Embedding
                       +
                Position Embedding
                       ↓
                 BERT Input Vector
```

### **The three ideas**

```text
Token Embedding
→ What token is this?

Segment Embedding
→ Which sentence/segment does it belong to?

Position Embedding
→ Where does it appear in the sequence?
```

---

# 📍 **Position Embeddings in BERT**

Unlike the original Transformer description that introduced fixed sinusoidal positional encoding, BERT uses **learned position embeddings**.

**Conceptually:**

```text
Token
  +
Position
  +
Segment
  ↓
Input Representation
```

For the BERT configurations discussed in these notes, the maximum sequence length is **512 tokens**.

```text
Maximum input token size
        ↓
       512
```

---

# 🔢 **Token Type / Segment IDs**

When BERT processes two sentences, `token_type_ids` indicate which segment each token belongs to.

**Example:**

```text
[CLS] First sentence [SEP] Second sentence [SEP]
  0       0            0      1       1       1
```

**Conceptually:**

```text
Segment 0 → First sentence
Segment 1 → Second sentence
```

For BERT-like models, Hugging Face tokenizers may return `token_type_ids` in addition to `input_ids` and `attention_mask`.

---

# 🔍 **Attention Mask**

The attention mask tells the model which positions are real tokens and which positions are padding.

**Example:**

```text
Input IDs:
[101, 1188, 1110, 1363, 102, 0, 0]

Attention Mask:
[ 1,    1,    1,    1,   1, 0, 0]
```

```text
1 → Real token
0 → Padding
```

This is the same tokenizer-level distinction used in the Hugging Face workflow notes.

---

# 🔤 **WordPiece Tokenization**

BERT commonly uses **WordPiece** tokenization.

A word can be split into smaller subword pieces.

**Example:**

```text

   ↓
play + ##ing
```

The idea of subword tokenization helps represent words using reusable pieces.

# ❓ **Question Answering**

Question Answering (QA) is another important BERT downstream task.

**The model receives:**

```text
Question + Context
```

and identifies the answer from the context.

**Typical BERT-style extractive QA predicts:**

```text
Start Position
+
End Position
```

**Conceptually:**

```text
Question + Context
        ↓
      BERT
        ↓
Contextual Representations
        ↓
 ┌───────────────┐
 ↓               ↓
Start Logits    End Logits
 ↓               ↓
Start Token     End Token
 └───────┬───────┘
         ↓
       Answer
```

---

# 📚 **Types of Question Answering**

**There are two useful categories to remember:**

## **1. Abstractive Question Answering**

> **Understand the context and generate an answer.**

```text
Context
  ↓
Understand
  ↓
Generate Answer
```

The generated answer does not have to be copied word-for-word from the context.

---

## **2. Extractive Question Answering**

> **Understand the context and extract the answer from it.**

```text
Context
  ↓
Find Answer Span
  ↓
Copy / Extract
```

In extractive QA, the answer is represented by a span from the supplied context.

---

# 🧪 **BERT Extractive QA Example**

### **Context**

```text
OpenAI is an American artificial intelligence (AI) research organization headquartered in San Francisco,
consisting of OpenAI Group PBC, a for-profit public benefit corporation (PBC),
partially controlled by OpenAI Foundation, a nonprofit.

OpenAI develops generative AI models, particularly the GPT series of large language models.

Its release of ChatGPT in November 2022 has been credited with catalyzing the AI boom,
and widespread interest in generative AI.
```

### **Question**

```text
When was ChatGPT released?
```

### **Expected Extractive Answer**

```text
November 2022
```

---



# 💻 **BERT Question Answering with Hugging Face**

````python
import torch

from transformers import (
    AutoTokenizer,
    AutoModelForQuestionAnswering,
)

# 1. Load the model and matching tokenizer
model_name = "deepset/bert-base-cased-squad2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

# 2. Setup inputs
context = "The Hugging Face Transformers library was released in 2018. It provides state-of-the-art NLP models."
question = "When was the Transformers library released?"

# 3. Tokenize question and context together
inputs = tokenizer(
    question,
    context,
    return_tensors="pt"
)

# 4. Perform model inference
with torch.no_grad():
    outputs = model(**inputs)

# 5. Extract start and end token span indices
answer_start = torch.argmax(outputs.start_logits)
answer_end = torch.argmax(outputs.end_logits) + 1

# 6. Decode the extracted token span
answer_tokens = inputs.input_ids[0][answer_start:answer_end]
answer = tokenizer.decode(answer_tokens)

print(f"Answer: {answer}")
````

---

# 🔬 **How the QA Code Works**

```text
Question + Context
        ↓
     Tokenizer
        ↓
input_ids + attention_mask
        ↓
BERT QA Model
        ↓
 ┌──────────────┐
 ↓              ↓
start_logits  end_logits
 ↓              ↓
start index   end index
 └──────┬───────┘
        ↓
   Answer Span
```

### **Important outputs**

```text
start_logits
→ Score for each possible starting token

end_logits
→ Score for each possible ending token
```

The highest-scoring start and end positions are then used to extract the answer span.

# ⚠️ **Bias & Limitations**

Masked language models learn patterns from their training data.

As a result, predictions can reflect patterns or biases present in that data.

**For example, when prompts differ only by demographic or social wording:**

```text
This man works as a [MASK].

This woman works as a [MASK].
```

the model may produce different candidate distributions.

**This is an important reminder:**

> **A pretrained language model does not automatically produce neutral or unbiased predictions.**

Outputs should be evaluated in context, especially for applications involving people, identity, employment, ranking, or decision-making.

# ✅ **BERT Quick Revision**

- **BERT** = Bidirectional Encoder Representations from Transformers.
- BERT is an **encoder-only Transformer**.
- BERT learns **bidirectional contextual representations**.
- The original pretraining setup used **MLM + NSP**.
- **MLM** predicts a masked token.
- **NSP** predicts whether one sentence follows another.
- `[CLS]` is used for classification-oriented representations.
- `[SEP]` separates sentence segments.
- BERT uses **token, segment, and learned position information**.
- The BERT configurations discussed here use a **maximum input length of 512 tokens**.
- **BERT Base** has about **110M parameters**, hidden size **768**, and **12 attention heads**.
- **BERT Large** has about **340M parameters**, hidden size **1024**, and **16 attention heads**.
- **Cased** models preserve capitalization.
- **Uncased** models do not preserve the capitalization distinction.
- BERT can be fine-tuned for **classification, token classification, question answering, and other NLP tasks**.
- **Extractive QA** finds an answer span in the context.
- **Abstractive QA** generates an answer from the understood context.
- BERT predictions can reflect **biases present in training data**.
