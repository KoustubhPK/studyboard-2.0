# 🧠 **Training Concepts**

Before working with Hugging Face, it is important to understand the difference between **Training**, **Transfer Learning**, and **Fine-Tuning**.

---

## 🏋️ **Training**

Training a model **from scratch** means:

> **The model architecture is created and its weights are initialized randomly.**

The model learns everything from the training data.

```text
Random Weights

      ↓

Training Data

      ↓

Forward Pass

      ↓

Loss

      ↓

Backpropagation

      ↓

Weight Update

      ↓

Learned Model
```

**For a model with parameters $\theta$, training tries to minimize the loss:**

$$
\theta^*
=
\arg\min_{\theta}
\mathcal{L}(\theta)
$$

**Training from scratch generally requires:**

- Large datasets
- Large computational resources
- Long training time
- Careful optimization
- Significant engineering effort

---

## 🔄 **Transfer Learning**

**Transfer Learning means:**

> **Using knowledge learned by a pretrained model for another task.**

**Instead of starting with random weights:**

```text
Random Weights
```

**we start with:**

```text
Pretrained Weights
```

**Example:**

```text
Pretrained BERT

        ↓

Sentiment Analysis
```

The pretrained model has already learned useful language representations.

---

## 🎯 **Fine-Tuning**

**Fine-Tuning means:**

> **Continuing the training of a pretrained model on a task-specific or domain-specific dataset.**

**The model starts with pretrained weights:**

```text
Pretrained Model

      ↓

Task-Specific Dataset

      ↓

Fine-Tuning

      ↓

Task-Specific Model
```

The weights are **initialized from the pretrained model**, not randomly.

**Therefore:**

```text
Training from Scratch
→ Random initialization

Fine-Tuning
→ Pretrained initialization
```

Fine-tuning usually requires much less compute, data, and time than pretraining a model from scratch.

# 📏 **Padding**

Sequences in a batch often have different lengths.

**Example:**

```text
Sentence 1:
I like transformers

Sentence 2:
I like transformers very much
```

Their lengths are different.

To create a batch, we can pad the shorter sequence.

```text
I like transformers <PAD> <PAD>

I like transformers very much
```

The padding tokens are ignored using the attention mask.

# 📈 **Logits**

A model usually produces **logits** before probabilities.

**For a classification model:**

```text
Input

 ↓

Model

 ↓

Logits

 ↓

Softmax

 ↓

Probabilities
```

**For $C$ classes:**

$$
\mathbf{z}
=
[z_1,z_2,\ldots,z_C]
$$

**Softmax converts logits into probabilities:**

$$
P(y=i)
=
\frac{e^{z_i}}
{\sum_{j=1}^{C}e^{z_j}}
$$

**The probabilities sum to:**

$$
\sum_{i=1}^{C}P(y=i)=1
$$

```python
import torch

text = "this is a good movie"

inputs = tokenizer(
    text,
    return_tensors="pt"
)

with torch.no_grad():
    outputs = model(**inputs)

print(outputs.logits)
```

# 🔄 **Dataset.map()**

The `map()` method is commonly used to apply preprocessing to every example.

**Example:**

```python
def tokenize_function(example):
    return tokenizer(
        example["text"],
        truncation=True
    )

tokenized_dataset = dataset.map(
    tokenize_function,
    batched=True
)
```

**Workflow:**

```text
Dataset

   ↓

map()

   ↓

Preprocessing Function

   ↓

Tokenized Dataset
```

---

# ⚡ **Why batched=True?**

**Without batching:**

```python
map(
    tokenize_function
)
```

the function processes examples individually.

**With:**

```python
map(
    tokenize_function,
    batched=True
)
```

the function receives a batch of examples.

This is often more efficient for tokenization.

---

# 🧹 **Removing Original Columns**

After tokenization, the original text column may no longer be required.

**Example:**

```python
tokenized_dataset = dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"]
)
```

# 🧠 **Trainer Workflow**

```text
Pretrained Model

      ↓

Training Dataset

      ↓

Tokenizer

      ↓

Data Collator

      ↓

TrainingArguments

      ↓

Trainer

      ↓

Fine-Tuned Model
```

```python
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

model_name = "bert-base-cased"

tokenizer = AutoTokenizer.from_pretrained(
    model_name
)

model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=2
)
```

**Create the trainer:**

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["test"],
    processing_class=tokenizer,
    data_collator=data_collator
)
```

**Train:**

```python
trainer.train()
```

# 📉 **Loss**

Training attempts to minimize a loss function.

For classification, cross-entropy loss is commonly used.

**For $C$ classes:**

$$
\mathcal{L}
=
-\sum_{i=1}^{C}
y_i\log(p_i)
$$

**where:**

- $y_i$ = true class indicator
- $p_i$ = predicted probability

For a single correct class $k$:

$$
\mathcal{L}
=
-\log(p_k)
$$

**Therefore:**

```text
Higher probability for correct class
        ↓
Lower loss

Lower probability for correct class
        ↓
Higher loss
```

---

# 🔁 **Backpropagation**

**Training follows the basic loop:**

```text
Input

 ↓

Model

 ↓

Prediction

 ↓

Loss

 ↓

Backpropagation

 ↓

Gradients

 ↓

Optimizer

 ↓

Updated Weights
```

**Mathematically:**

$$
\theta
\leftarrow
\theta
-
\eta
\nabla_\theta\mathcal{L}
$$

**where:**

- $\theta$ = model parameters
- $\eta$ = learning rate
- $\nabla_\theta\mathcal{L}$ = gradient of loss with respect to parameters

---

# 📐 **Learning Rate**

The learning rate controls how large the parameter updates are.

```text
Very Small Learning Rate
→ Slow Training

Very Large Learning Rate
→ Training may become unstable

Good Learning Rate
→ Efficient Optimization
```

For fine-tuning pretrained models, learning rates are often smaller than those used when training a model from scratch.

---

# 📦 **Batch Size**

Batch size is the number of examples processed together.

**Example:**

```python
per_device_train_batch_size=8
```

**means:**

```text
8 examples
   ↓
Forward Pass
   ↓
Loss
   ↓
Backward Pass
   ↓
Update
```

Larger batch sizes generally require more memory.

---

# 🔢 **Gradient Accumulation**

If GPU memory cannot hold a large batch, gradient accumulation can simulate a larger effective batch.

**Example:**

```python
training_args = TrainingArguments(
    output_dir="model",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4
)
```

**Approximate effective batch size:**

$$
\text{Effective Batch Size}
=
\text{Batch Size}
\times
\text{Gradient Accumulation Steps}
$$

**Therefore:**

$$
2\times4=8
$$

# ⚖️ **Full Fine-Tuning vs PEFT**

## **Full Fine-Tuning**

```text
Pretrained Model

      ↓

Update Most / All Model Parameters

      ↓

Fine-Tuned Model
```

**Advantages:**

- Maximum flexibility
- Can adapt all model parameters

**Disadvantages:**

- High GPU memory usage
- Large optimizer state

- Expensive for large models

## **PEFT / LoRA**

```text
Pretrained Model
      |
      | Freeze Base Model
      ↓
LoRA Adapter
      |
      ↓
Train Small Number of Parameters
```

PEFT methods train a much smaller number of parameters than full fine-tuning.

---

# 🧩 **LoRA**

**LoRA = Low-Rank Adaptation**

Instead of directly updating a large weight matrix $W$, LoRA learns a low-rank update.

**Conceptually:**

$$
W'
=
W
+
\Delta W
$$

**LoRA represents the update as:**

$$
\Delta W
=
BA
$$

**where:**

```text
A → Small Trainable Matrix

B → Small Trainable Matrix
```

The pretrained matrix $W$ can remain frozen.

**Therefore:**

```text
Original Weight W

      +

Small Trainable Update BA

      ↓

Adapted Weight
```

---

# 📐 **LoRA Parameter Reduction**

**Suppose:**

$$
W\in\mathbb{R}^{d\times d}
$$

**Full fine-tuning requires approximately:**

$$
d^2
$$

trainable parameters.

**LoRA with rank $r$ uses:**

$$
2dr
$$

trainable parameters.

**If:**

$$
r\ll d
$$

**then:**

$$
2dr \ll d^2
$$

This is why LoRA can dramatically reduce the number of trainable parameters.

# 🌡️ **Temperature**

Temperature controls the sharpness of the probability distribution during sampling.

**Conceptually:**

```text
Low Temperature
→ More deterministic

High Temperature
→ More diverse / random
```

**For logits $z_i$:**

$$
P_i
=
\frac{
e^{z_i/T}
}{
\sum_j e^{z_j/T}
}
$$

where $T$ is temperature.

```python
outputs = model.generate(
    **inputs,
    max_new_tokens=30,
    do_sample=True,
    top_p=0.9
)
```

Top-k and top-p are often used with sampling-based generation.

# 🏁 **Final Summary**

- **Hugging Face** provides an ecosystem for building and sharing modern ML and AI applications.
- The **Hugging Face Hub** provides access to models, datasets, and Spaces.
- **Transformers** is the main library for working with pretrained Transformer models.
- **Pipeline** provides a high-level abstraction for inference.
- **AutoTokenizer** converts raw text into the representation expected by a model.
- `input_ids` contain token IDs.
- `attention_mask` identifies real tokens versus padding.
- `padding` makes sequences in a batch compatible with one another.
- `truncation` limits sequences to a specified maximum length.
- **AutoModel classes** provide task-specific model loading.
- **Datasets** provides tools for loading and processing datasets.
- `dataset.map()` is commonly used for preprocessing and tokenization.
- **Trainer** provides a high-level training and evaluation loop.
- **TrainingArguments** controls important training settings.
- **Fine-tuning** starts from pretrained weights rather than random initialization.
- **PEFT** reduces the number of trainable parameters required for fine-tuning.
- **LoRA** learns low-rank updates while keeping the base model largely frozen.
- **QLoRA** combines 4-bit quantization with LoRA.
- **bitsandbytes** provides common 4-bit and 8-bit quantization workflows.
- **Accelerate** helps scale PyTorch training across hardware and distributed environments.
- **TRL** provides post-training trainers such as SFT, DPO, and GRPO.
- **Gradio** can be used to create simple ML interfaces.
- **Evaluate** provides common evaluation metrics.
- **huggingface_hub** provides programmatic access to the Hugging Face Hub.
- Models and tokenizers can be saved with `save_pretrained()`.
- Fine-tuned models can be uploaded using `push_to_hub()`.
- A useful overall mental model is:

```text
Dataset

   ↓

Preprocess

   ↓

Tokenizer

   ↓

Pretrained Model

   ↓

Inference / Fine-Tuning

   ↓

Evaluation

   ↓

Save

   ↓

Hugging Face Hub
```
