# __Gated Recurrent Unit (GRU)__

Before understanding **GRU**, let us quickly recall the limitations of previous neural network architectures for sequential data.

---

# 🧠 __1. Artificial Neural Network (ANN)__

A standard **Artificial Neural Network (ANN)** assumes that all input features are independent of one another.

While this works well for many structured datasets, it is not suitable for sequential data such as text, speech, or time series.

## __Limitations of ANN on Text Data__

- Does **not preserve the order** of words.
- Cannot naturally handle **variable-length input and output sequences**.

---

# 🔄 __2. Recurrent Neural Network (RNN)__

A **Recurrent Neural Network (RNN)** was introduced to process sequential data by maintaining a hidden state that carries information from previous time steps.

## __Benefits of RNN__

- Preserves the **order** of sequence data by processing one time step at a time.
- Can handle **variable-length input and output sequences** through parameter sharing.

---

## __Limitations of RNN__

Although RNNs are designed for sequential data, they have two major drawbacks.

- Unable to effectively process **long sequences** due to the Vanishing Gradient Problem.
- Unable to capture **future context**, since information flows only from past to future.

---

# 🧠 __3. Long Short-Term Memory (LSTM)__

LSTM was introduced to overcome the inability of Vanilla RNNs to learn long-term dependencies.

## __Benefits of LSTM__

- Can learn and preserve **long-term dependencies**.
- Greatly reduces the **Vanishing Gradient Problem**.

---

## __Limitations of LSTM__

Although LSTM performs much better than a Vanilla RNN, it still has some disadvantages.

- Unable to capture **future context**.
- More complex architecture.
- Computationally expensive.
- Requires longer training time.

---

# 🚀 __4. Why GRU Was Introduced__

The **Gated Recurrent Unit (GRU)** was introduced as a simpler alternative to the LSTM.

It retains the ability to learn long-term dependencies while reducing the complexity of the architecture.

Compared to an LSTM, a GRU

- Uses fewer gates.
- Maintains only one state.
- Has fewer parameters.
- Trains faster.
- Requires less memory.

---

# 🧠 __5. Gated Recurrent Unit (GRU)__

A **Gated Recurrent Unit (GRU)** is a special type of **Recurrent Neural Network (RNN)** that improves the ability of a Vanilla RNN to learn long-term dependencies.

Unlike an LSTM, a GRU does **not** maintain a separate Cell State.

Instead, it uses only a **Hidden State** to store both short-term and long-term information.

---

# 🔄 __6. Difference Between LSTM and GRU__

| LSTM | GRU |
|------|-----|
| Has **2 states**: Hidden State and Cell State | Has only **1 state**: Hidden State |
| Uses **3 gates** | Uses **2 gates** |
| Input Gate, Forget Gate, Output Gate | Update Gate, Reset Gate |
| More parameters | Fewer parameters |
| More computationally expensive | Less computationally expensive |
| Slower training | Faster training |
| More complex architecture | Simpler architecture |

---

# 🚪 __7. Gates in GRU__

A GRU uses only **two gates** to control the flow of information.

- **Update Gate**
- **Reset Gate**

These gates decide

- What information should be remembered.
- What information should be forgotten.

Like an LSTM, both gates use the **Sigmoid Activation Function**, whose output lies between **0 and 1**.

---

# 🔄 __8. Update Gate__

The **Update Gate** determines **how much new information should be added** to the Hidden State.

It controls how much of the previous Hidden State should be retained and how much new information should be incorporated.

### __Intuition__

The Update Gate answers the question:

> **"What new information should I remember?"**

A high Update Gate value means

- Preserve important information from previous time steps.
- Update the Hidden State only where necessary.

---

# 🔄 __9. Reset Gate__

The **Reset Gate** determines **how much information from the previous Hidden State should be ignored** before computing the new Hidden State.

### __Intuition__

The Reset Gate answers the question:

> **"What old information should I forget?"**

A low Reset Gate value causes irrelevant past information to be discarded before processing the current input.

---

# 🔄 __10. Working of a GRU__

At every time step, the GRU performs the following operations.

### __Step 1__

Receive

- Current Input ($x_t$)
- Previous Hidden State ($h_{t-1}$)

---

### __Step 2__

The **Update Gate** decides how much previous information should be retained.

---

### __Step 3__

The **Reset Gate** determines which previous information should be ignored.

---

### __Step 4__

A new Hidden State is computed using the current input and the gated previous Hidden State.

---

### __Step 5__

The updated Hidden State ($h_t$) is passed to the next time step and is also used for generating the output.

---

# 📌 __11. Advantages of GRU__

- Simpler architecture than LSTM.
- Fewer parameters.
- Faster training.
- Lower memory consumption.
- Can learn long-term dependencies.
- Performs well on many sequential learning tasks.

---

# ⚠️ __12. Limitations of GRU__

- Cannot capture future context.
- Slightly less expressive than LSTM for some complex tasks.
- Still processes sequences sequentially, making parallelization difficult.

---

# 🔄 __13. Difference Between LSTM and GRU__

| LSTM | GRU |
|------|-----|
| Hidden State + Cell State | Hidden State only |
| 3 Gates | 2 Gates |
| More parameters | Fewer parameters |
| More complex | Simpler |
| Slower training | Faster training |
| Higher memory usage | Lower memory usage |

---

# 📝 __Summary__

- GRU is a simplified version of LSTM.
- It combines long-term and short-term memory into a single Hidden State.
- It uses only two gates: the Update Gate and the Reset Gate.
- Due to fewer parameters, GRUs train faster while still learning long-term dependencies effectively.
- GRUs are widely used for text classification, language modeling, speech recognition, and time-series forecasting.

# __Bidirectional Recurrent Neural Network (Bidirectional RNN)__

A standard RNN, LSTM, or GRU processes a sequence only in the **forward direction**.

This means that the prediction at each time step depends only on the **past context**.

However, in many applications, future words also provide useful information.

To overcome this limitation, **Bidirectional RNNs** were introduced.

---

# ⚠️ __1. Limitation of Unidirectional RNN__

A standard RNN processes a sequence only from **left to right**.

For example,

> **Apple is a great company.**

While processing the word **Apple**, the model has not yet seen the word **company**.

Therefore, it may not correctly determine whether **Apple** refers to

- the fruit, or
- the company.

---

# 🚀 __2. Why Bidirectional RNN?__

A Bidirectional RNN processes the sequence in **both directions**.

- Forward direction → Past → Future
- Backward direction → Future → Past

This allows each time step to use information from both the past and the future.

---

# 🏗️ __3. Architecture of Bidirectional RNN__

A Bidirectional RNN consists of two separate recurrent networks.

- Forward RNN
- Backward RNN

Their outputs are combined before making the final prediction.

# 🍎 __4. Example__

Consider the sentence

> **Apple is a great company.**

### __Forward Processing__

```
h₁ = Apple

h₂ = Apple is

h₃ = Apple is great

h₄ = Apple is great company
```

If we predict the entity of **Apple** using only the forward hidden state,

```
Entity of Apple → h₁ → Output Layer
```

The model has only seen the word **Apple**, so it may incorrectly classify it as a fruit.

---

### __Backward Processing__

```
h₄' = Company

h₃' = Great Company

h₂' = Is Great Company

h₁' = Apple Is Great Company
```

Now, the hidden state corresponding to **Apple** also contains information from the future.

```
Entity of Apple → h₁ + h₁' → Output Layer
```

Since the backward network has already seen the word **company**, the model can correctly classify **Apple** as an organization.

---

# ✅ __5. Advantages of Bidirectional RNN__

- Uses both past and future context.
- Produces better contextual representations.
- Improves performance on many NLP tasks.
- Can be combined with RNN, LSTM, or GRU.

---

# ⚠️ __6. Limitations of Bidirectional RNN__

- Computationally more expensive.
- Requires more memory.
- Cannot be used for real-time prediction where future inputs are unavailable.

---

# 💻 __7. Bidirectional Layers in TensorFlow__

```python
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, Dense, SimpleRNN, GRU
from tensorflow.keras.models import Sequential

model = Sequential()

# Bidirectional Simple RNN
model.add(Bidirectional(SimpleRNN(20)))

# Bidirectional LSTM
model.add(Bidirectional(LSTM(10)))

# Bidirectional GRU
model.add(Bidirectional(GRU(10)))
```

---

# 📝 __Summary__

- A Bidirectional RNN processes sequences in both forward and backward directions.
- It captures both past and future context.
- It improves sequence understanding and prediction accuracy.
- Bidirectional wrappers can be used with Simple RNN, LSTM, and GRU layers.
- They are commonly used in tasks such as Named Entity Recognition (NER), Machine Translation, Part-of-Speech Tagging, and Sentiment Analysis.
