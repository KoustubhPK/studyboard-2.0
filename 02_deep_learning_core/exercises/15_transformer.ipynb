# **Transformer**

The **Transformer** architecture was introduced in **2017** by researchers at **Google** in the paper **"Attention Is All You Need"**.

The Transformer was built on the idea of **Attention Mechanism** and removed the need for recurrent processing used by traditional RNN-based Encoder-Decoder architectures.

---

# 🚀 **Evolution of Sequence-to-Sequence Models**

| Year | Development |
|---|---|
| **2014 (Google)** | Encoder-Decoder Architecture |
| **2015** | Attention Mechanism |
| **2017** | Transformer |

# 🎯 **Task : Machine Translation**

We will understand the Transformer using the task of **Machine Translation**.

Suppose we have:

```text
"We are learning transformer"
```

and we want to translate it into Hindi:

```text
"Hum transformer sikh rahe hai"
```

The Transformer consists of two major components:

```text
Encoder : Process the source language

Decoder : Generate the target language
```

---

# 🏗️ **Transformer Architecture**

```text
                Transformer
                     |
          -----------------------
          |                     |
       Encoder               Decoder
          |                     |
   Source Language        Target Language
      (English)               (Hindi)
```

The **Encoder** understands the input sentence.

The **Decoder** generates the output sentence one token at a time.

---

# 🔵 **Encoder**

The Encoder is responsible for:

> **Processing and understanding the source language.**

Our input sentence is:

```text
"We are learning transformer"
```

---

## **Step 1 — Tokenization**

The sentence is first divided into tokens.

```text
"We are learning transformer"

↓

["We", "are", "learning", "transformer"]
```

---

## **Step 2 — Input IDs**

Every token is mapped to an integer ID using the tokenizer vocabulary.

For example:

```text
"We"          --> 876

"are"         --> 10

"learning"    --> 56

"transformer" --> 213
```

Therefore:

```text
Input IDs

[876, 10, 56, 213]
```

> The actual IDs depend on the tokenizer and vocabulary.

---

## **Step 3 — Input Embedding**

The Input IDs are converted into dense vectors using an **Embedding Layer**.

```text
Input IDs

[876, 10, 56, 213]

        ↓

Embedding Layer

        ↓

W, A, L, T

        ↓

Embedding Vectors
```

For example:

```text
876 --> Vector for "We"

10  --> Vector for "are"

56  --> Vector for "learning"

213 --> Vector for "transformer"
```

The embedding vectors contain semantic information about the tokens.

---

# 📐 **Embedding Dimension**

In the original Transformer paper:

```text
d_model = 512
```

Therefore, each token is represented using a vector of dimension:

```text
512
```

For example:

```text
"We"

↓

[0.12, -0.43, 0.81, ..., 0.27]

        512 dimensions
```

---

## 📍 **Step 4 — Positional Encoding**

Self-Attention does not inherently understand the order of tokens.

Consider:

```text
Cat chased dog.
```

and:

```text
Dog chased cat.
```

The same words are present, but the meaning is different because the order of the words is different.

Therefore:

> **Language is sensitive to the order of tokens.**

---

# 🧭 **Why Positional Encoding?**

Self-Attention processes tokens in parallel.

Therefore, the Transformer needs additional information to know:

```text
Which token comes first?

Which token comes second?

Which token comes third?

...
```

This information is provided using **Positional Encoding**.

---

## **Positions**

For example:

```text
1st position --> P₁

2nd position --> P₂

3rd position --> P₃

4th position --> P₄
```

The positional encoding vector has the same dimension as the token embedding.

```text
Dim(Positional Encoding)

=

Dim(Input Embedding)

=

d_model
```

In the original Transformer:

```text
d_model = 512
```

---

## ➕ **Adding Positional Encoding**

The token embedding and positional encoding are added element-wise.

For example:

```text
"We"          --> W

"are"         --> A

"learning"    --> L

"transformer" --> T
```

After adding positional information:

```text
W + P₁

A + P₂

L + P₃

T + P₄
```

Therefore:

```text
W, A, L, T

      +

P₁, P₂, P₃, P₄

      ↓

W+P₁, A+P₂, L+P₃, T+P₄
```

These vectors are then passed to the Encoder layers.

---

# 📌 **Positional Encoding in the Original Transformer**

The original Transformer paper used **sinusoidal positional encoding**.

The equations are:

$$
PE_{(pos,2i)}
=
\sin
\left(
\frac{pos}
{10000^{\frac{2i}{d_{model}}}}
\right)
$$

$$
PE_{(pos,2i+1)}
=
\cos
\left(
\frac{pos}
{10000^{\frac{2i}{d_{model}}}}
\right)
$$

where:

- $pos$ = position of the token
- $i$ = dimension index
- $d_{model}$ = embedding dimension

In the original Transformer:

```text
d_model = 512
```

---

# 🧠 **Encoder Layer**

The Transformer Encoder contains multiple Encoder layers.

In the original Transformer:

```text
Number of Encoder Layers = 6
```

Each Encoder layer contains:

```text
Input

↓

Multi-Head Self-Attention

↓

Add & Norm

↓

Feed Forward Network

↓

Add & Norm

↓

Output
```

Therefore:

```text
W+P₁, A+P₂, L+P₃, T+P₄

        ↓

Multi-Head Self-Attention

        ↓

Add & Norm

        ↓

Feed Forward Network

        ↓

Add & Norm

        ↓

Output
```

The output of one Encoder layer becomes the input to the next Encoder layer.

---

# 🧠 **Multi-Head Self-Attention**

The first major component of the Encoder is:

> **Multi-Head Self-Attention**

Input to Multi-Head Attention:

```text
W+P₁

A+P₂

L+P₃

T+P₄
```

These vectors are converted into:

- Query (Q)
- Key (K)
- Value (V)

using learnable weight matrices.

---

# **Query, Key and Value**

For every input vector:

```text
Input

↓

Linear Layers

↓

Query (Q)

Key (K)

Value (V)
```

For example:

```text
W+P₁

↓

Wq, Wk, Wv
```

Similarly:

```text
A+P₂

↓

Aq, Ak, Av
```

```text
L+P₃

↓

Lq, Lk, Lv
```

```text
T+P₄

↓

Tq, Tk, Tv
```

---

# ❓ **Why Query, Key and Value?**

Suppose we want to understand the word:

```text
"learning"
```

The Query represents:

> **What information am I looking for?**

The Key represents:

> **What information does each token contain?**

The Value represents:

> **What information should actually be passed forward?**

Therefore:

```text
Query

↓

Asks the question


Key

↓

Determines relevance


Value

↓

Provides the information
```

---

# 🔢 **Attention Calculation**

Each Query is compared with every Key.

For example, for the token **"learning"**:

```text
Lq · Wk

Lq · Ak

Lq · Lk

Lq · Tk
```

These dot products produce attention scores.

The complete process is:

```text
Query + Key

↓

Dot Product

↓

Attention Scores

↓

Scaling

↓

Softmax

↓

Attention Weights

↓

Weighted Sum of Values

↓

Attention Output
```

---

# 📐 **Scaled Dot-Product Attention**

The attention mechanism is calculated using:

$$
\text{Attention}(Q,K,V)
=
\text{Softmax}
\left(
\frac{QK^\top}{\sqrt{d_k}}
\right)V
$$

where:

- $Q$ = Query matrix
- $K$ = Key matrix
- $V$ = Value matrix
- $d_k$ = Dimension of Key vectors

---

# **Step 1 — Dot Product**

The Query is compared with every Key.

For example:

```text
Lq · Wk

Lq · Ak

Lq · Lk

Lq · Tk
```

These values represent the relevance between the Query and each token.

```text
Higher Score

↓

Higher Relevance
```

---

# **Step 2 — Scaling**

The dot-product scores are divided by:

$$
\sqrt{d_k}
$$

Therefore:

$$
\text{Scaled Score}
=
\frac{QK^\top}{\sqrt{d_k}}
$$

Scaling prevents the dot-product values from becoming too large when the dimension of the Key vectors is large.

This helps keep the Softmax function in a useful range and improves training stability.

---

# **Step 3 — Softmax**

The scaled attention scores are passed through the Softmax function.

```text
Scaled Scores

↓

Softmax

↓

Attention Weights
```

The attention weights represent:

> **How much attention should be given to each token.**

For example:

```text
Token          Attention Weight

"We"           → 0.10

"are"          → 0.15

"learning"     → 0.60

"transformer"  → 0.15
```

The values sum to:

$$
\sum_i P_i = 1
$$

---

# **Step 4 — Weighted Sum of Values**

The attention weights are multiplied with the corresponding Value vectors.

For example:

$$
Z_1
=
P_{11}W_v
+
P_{12}A_v
+
P_{13}L_v
+
P_{14}T_v
$$

Similarly, every token produces its own attention output.

Therefore:

```text
"We"          → Z₁

"are"         → Z₂

"learning"    → Z₃

"transformer" → Z₄
```

These outputs contain contextual information about each token.

---

# 🔄 **Why Self-Attention?**

It is called **Self-Attention** because:

> Query, Key and Value are generated from the **same input sequence**.

For example:

```text
Input Sequence

"We are learning transformer"

        ↓

      Query
      Key
      Value

        ↓

Self-Attention
```

Every token can attend to every other token in the same sequence.

---

# 🧠 **Multi-Head Attention**

Instead of performing Attention only once, the Transformer performs Attention multiple times in parallel.

These are called:

> **Attention Heads**

For example:

```text
Head 1

Head 2

Head 3

Head 4

...

Head H
```

In the original Transformer:

```text
Number of Heads = 8
```

---

# ❓ **Why Multiple Heads?**

A sentence can contain different types of relationships.

For example:

```text
"Tim Cook is CEO of Apple."
```

The model may need to understand:

- Semantic relationships
- Syntactic relationships
- Grammatical relationships
- Subject-object relationships
- Long-range dependencies
- Word references

Different attention heads can learn different relationships.

For example:

```text
Head 1 → Semantic relationship

Head 2 → Syntactic relationship

Head 3 → Grammatical relationship

Head 4 → Long-range dependency
```

> The exact behavior of each attention head is learned automatically during training.

---

# 🔀 **Different Weight Matrices for Different Heads**

Although every attention head receives the same input, every head has its own learnable weight matrices.

For Head 1:

```text
Wq₁

Wk₁

Wv₁
```

For Head 2:

```text
Wq₂

Wk₂

Wv₂
```

Similarly:

```text
Wq₃, Wk₃, Wv₃

...

Wqₕ, Wkₕ, Wvₕ
```

Therefore, every head can learn a different representation of the same input.

---

# 🧮 **Attention in Each Head**

For Head $i$:

$$
head_i
=
\text{Attention}
\left(
QW_i^Q,
KW_i^K,
VW_i^V
\right)
$$

where:

- $W_i^Q$ = Query projection matrix for Head $i$
- $W_i^K$ = Key projection matrix for Head $i$
- $W_i^V$ = Value projection matrix for Head $i$

Each head performs:

```text
Query

↓

Key

↓

Dot Product

↓

Scaling

↓

Softmax

↓

Weighted Sum of Values

↓

Head Output
```

---

# 🔗 **Concatenation of Attention Heads**

Suppose we have $H$ attention heads.

Each head produces an output:

```text
Head 1 → Z₁

Head 2 → Z₂

Head 3 → Z₃

...

Head H → Zₕ
```

These outputs are concatenated:

$$
\text{Concat}
\left(
head_1,
head_2,
\ldots,
head_H
\right)
$$

Then a final linear projection is applied:

$$
\text{MultiHead}(Q,K,V)
=
\text{Concat}
\left(
head_1,
head_2,
\ldots,
head_H
\right)
W^O
$$

where:

- $W^O$ = Output projection matrix

Therefore:

```text
Head 1 ──┐
Head 2 ──┤
Head 3 ──┤
   ...    ├──→ Concatenate
Head H ──┘
              ↓
        Linear Projection
              ↓
      Multi-Head Attention
           Output
```

---

# ➕ **Add & Norm**

After Multi-Head Attention, the Transformer applies:

```text
Add & Norm
```

It consists of:

```text
Add

↓

Residual Connection


Norm

↓

Layer Normalization
```

Therefore:

```text
Multi-Head Attention Output

          +

Original Input

          ↓

Residual Connection

          ↓

Layer Normalization

          ↓

Output
```

---

# ➕ **Add : Residual Connection**

A Residual Connection adds the input of the component to its output.

Input to Multi-Head Attention:

```text
W+P₁

A+P₂

L+P₃

T+P₄
```

Output of Multi-Head Attention:

```text
Z₁

Z₂

Z₃

Z₄
```

The residual connection performs:

```text
Z₁ + (W+P₁)

Z₂ + (A+P₂)

Z₃ + (L+P₃)

Z₄ + (T+P₄)
```

In matrix form:

$$
X + \text{MHA}(X)
$$

where:

- $X$ = Input to Multi-Head Attention
- $\text{MHA}(X)$ = Output of Multi-Head Attention

---

# ❓ **Why Residual Connection?**

Residual connections help:

- Preserve the original information of the input.
- Improve information flow.
- Improve gradient flow.
- Make deep networks easier to train.
- Help reduce the vanishing-gradient problem.

Conceptually:

```text
Input
  │
  ├─────────────────────┐
  │                     ↓
  │              Multi-Head
  │              Attention
  │                     │
  │                     ↓
  └──────────────────→ Add
                        ↓
                      Output
```

---

# 📏 **Norm : Layer Normalization**

After the residual addition, **Layer Normalization** is applied.

```text
Residual Output

      ↓

Layer Normalization

      ↓

Normalized Output
```

Mathematically:

$$
X'
=
\text{LayerNorm}
\left(
X + \text{MHA}(X)
\right)
$$

---

# ❓ **Why Layer Normalization?**

Layer Normalization helps:

- Stabilize activations.
- Improve training stability.
- Make optimization easier.
- Allow deep Transformer networks to train effectively.

The main purpose is to keep the activations well-behaved during training.

---

# 🧠 **Feed Forward Network**

After Multi-Head Attention and Add & Norm, the output is passed through a:

> **Feed Forward Network**

The Feed Forward Network is also called:

> **Position-wise Feed Forward Network**

It is applied independently to every token.

For example:

```text
Token 1 → Feed Forward

Token 2 → Feed Forward

Token 3 → Feed Forward

Token 4 → Feed Forward
```

The same Feed Forward Network is applied to every position.

---

# **Feed Forward Network**

The original Transformer uses two linear transformations with a **ReLU** activation function.

The equation is:

$$
FFN(x)
=
\text{ReLU}
\left(
xW_1+b_1
\right)
W_2+b_2
$$

where:

- $W_1$ = First linear layer weights
- $b_1$ = First linear layer bias
- $W_2$ = Second linear layer weights
- $b_2$ = Second linear layer bias

---

# 📐 **Feed Forward Dimensions**

In the original Transformer:

```text
d_model = 512

d_ff = 2048
```

Therefore:

```text
512

↓

Linear Layer

↓

2048

↓

ReLU

↓

Linear Layer

↓

512
```

The dimensional transformation is:

$$
512
\rightarrow
2048
\rightarrow
512
$$

---

# ❓ **Why Feed Forward Network?**

Self-Attention allows tokens to interact with one another.

The Feed Forward Network applies a non-linear transformation to each token representation.

Therefore:

```text
Multi-Head Attention

↓

Token-to-Token Interaction

↓

Feed Forward Network

↓

Non-Linear Transformation
```

The Feed Forward Network helps the model learn complex patterns and transformations.

---

# ➕ **Second Add & Norm**

After the Feed Forward Network, another Residual Connection and Layer Normalization are applied.

```text
Feed Forward Output

        +

Input to Feed Forward

        ↓

Residual Connection

        ↓

Layer Normalization

        ↓

Output of Encoder Layer
```

Mathematically:

$$
Y
=
\text{LayerNorm}
\left(
X'
+
FFN(X')
\right)
$$

where:

$$
X'
=
\text{LayerNorm}
\left(
X
+
\text{MHA}(X)
\right)
$$

Therefore, one complete Encoder layer is:

```text
Input

↓

Multi-Head Self-Attention

↓

Add & Norm

↓

Feed Forward Network

↓

Add & Norm

↓

Output
```

---

# 🔁 **Stacking Encoder Layers**

The original Transformer contains:

```text
6 Encoder Layers
```

The output of one Encoder layer becomes the input to the next Encoder layer.

```text
Input Embedding
      +
Positional Encoding

        ↓

Encoder Layer 1

        ↓

Encoder Layer 2

        ↓

Encoder Layer 3

        ↓

Encoder Layer 4

        ↓

Encoder Layer 5

        ↓

Encoder Layer 6

        ↓

Encoder Output
```

Each Encoder layer produces richer contextual representations.

---

# 📤 **Output of Encoder**

The Encoder produces a contextual representation for every input token.

Input:

```text
"We are learning transformer"
```

↓

```text
W+P₁

A+P₂

L+P₃

T+P₄
```

↓

```text
Encoder

↓

V₁

V₂

V₃

V₄
```

Therefore:

```text
"We"          → V₁

"are"         → V₂

"learning"    → V₃

"transformer" → V₄
```

These vectors contain contextual information about the source sentence.

> The Encoder produces a sequence of contextual representations, not a single context vector.

---

# 🟢 **Decoder**

The Decoder is responsible for:

> **Generating the target language.**

Our target language is Hindi:

```text
"Hum transformer sikh rahe hai"
```

The Decoder generates the target sentence **one token at a time**.

---

# 🤖 **Autoregressive Generation**

The Decoder generates the output sequence autoregressively.

> **Autoregressive means that the next token is generated using the previously generated tokens.**

For example:

```text
<SoS>

↓

Hum

↓

transformer

↓

sikh

↓

rahe

↓

hai

↓

<EoS>
```

The output of the previous step becomes part of the input for the next step.

---

# 🏁 **Special Tokens**

The Decoder uses special tokens to identify the beginning and end of the sequence.

```text
<SoS> → Start of Sequence

<EoS> → End of Sequence
```

The generation starts with:

```text
<SoS>
```

and stops when:

```text
<EoS>
```

is generated.

---

# ⏱️ **Autoregressive Generation Example**

Target sentence:

```text
"Hum transformer sikh rahe hai"
```

The generation process is:

| Time Step | Input to Decoder | Output |
|---|---|---|
| $t=0$ | `<SoS>` | `Hum` |
| $t=1$ | `<SoS> Hum` | `transformer` |
| $t=2$ | `<SoS> Hum transformer` | `sikh` |
| $t=3$ | `<SoS> Hum transformer sikh` | `rahe` |
| $t=4$ | `<SoS> Hum transformer sikh rahe` | `hai` |
| $t=5$ | `<SoS> Hum transformer sikh rahe hai` | `<EoS>` |

Therefore:

```text
Input to Decoder

<SoS>

        ↓

Output

Hum

        ↓

Input to next step

<SoS> Hum
```

and the process continues.

---

# 📥 **Decoder Input Embedding**

The target tokens are also converted into embeddings.

For example:

```text
<SoS>

Hum

transformer

sikh

rahe

hai
```

↓

```text
Token IDs

↓

Embedding Layer

↓

Embedding Vectors
```

Just like the Encoder, positional information is also added.

```text
Target Embeddings

        +

Positional Encoding

        ↓

Decoder Input
```

---

# 🔒 **Masked Multi-Head Self-Attention**

The first component of the Decoder is:

> **Masked Multi-Head Self-Attention**

The Decoder must not look at future tokens while predicting the current token.

For example, when predicting:

```text
"transformer"
```

the Decoder can use:

```text
<SoS> Hum
```

but it cannot use:

```text
sikh rahe hai
```

because those are future tokens.

---

# ❓ **Why Masking?**

Suppose the target sentence is:

```text
Hum transformer sikh rahe hai
```

When predicting:

```text
sikh
```

the Decoder should only see:

```text
<SoS> Hum transformer
```

It should not see:

```text
rahe hai
```

Therefore, future tokens are masked.

```text
Current Token

        ↓

Can attend to:

Previous Tokens
+
Current Token

        ↓

Cannot attend to:

Future Tokens
```

---

# 🔺 **Causal Mask**

The mask prevents a token from attending to future tokens.

For example:

```text
              <SoS>   Hum   transformer   sikh   rahe   hai

<SoS>          ✓       ✕         ✕          ✕      ✕      ✕

Hum            ✓       ✓         ✕          ✕      ✕      ✕

transformer    ✓       ✓         ✓          ✕      ✕      ✕

sikh           ✓       ✓         ✓          ✓      ✕      ✕

rahe           ✓       ✓         ✓          ✓      ✓      ✕

hai            ✓       ✓         ✓          ✓      ✓      ✓
```

This is called:

> **Causal Mask** or **Look-Ahead Mask**

---

# 🧠 **Masked Attention**

The attention calculation itself remains the same:

$$
\text{Attention}(Q,K,V)
=
\text{Softmax}
\left(
\frac{QK^\top}{\sqrt{d_k}}
\right)V
$$

The difference is that the future positions are masked before applying Softmax.

Conceptually:

```text
Attention Scores

        ↓

Apply Causal Mask

        ↓

Softmax

        ↓

Attention Weights

        ↓

Weighted Sum of Values
```

---

# 🔗 **Cross-Attention**

After Masked Self-Attention, the Decoder performs:

> **Cross-Attention**

Cross-Attention allows the Decoder to look at the Encoder output.

The Encoder has already processed:

```text
"We are learning transformer"
```

and produced:

```text
V₁

V₂

V₃

V₄
```

These Encoder outputs are used by the Decoder.

---

# 🔄 **Query, Key and Value in Cross-Attention**

In Cross-Attention:

```text
Query

↓

Comes from Decoder
```

while:

```text
Key

Value

↓

Come from Encoder
```

Therefore:

```text
Decoder Output

↓

Query
```

and:

```text
Encoder Output

↓

Key + Value
```

This is different from Self-Attention.

---

# 🔑 **Self-Attention vs Cross-Attention**

## **Self-Attention**

Query, Key and Value come from the **same sequence**.

```text
Sequence

↓

Q, K, V

↓

Self-Attention
```

Examples:

```text
Encoder Self-Attention

Decoder Masked Self-Attention
```

---

## **Cross-Attention**

Query comes from one sequence.

Key and Value come from another sequence.

```text
Decoder

↓

Query


Encoder

↓

Key + Value

        ↓

Cross-Attention
```

Therefore:

```text
Q → Decoder

K → Encoder

V → Encoder
```

---

# 🌉 **Encoder-Decoder Connection**

The complete connection between Encoder and Decoder is:

```text
Source Language

"We are learning transformer"

        ↓

      Encoder

        ↓

Encoder Output

V₁ V₂ V₃ V₄

        ↓
        |
        | Key + Value
        |
        ↓

   Cross-Attention

        ↑
        |
        | Query
        |
      Decoder
```

The Decoder uses the Encoder representations to understand which source tokens are relevant while generating the target sentence.

---

# 🧠 **Example of Cross-Attention**

Suppose the Decoder is generating:

```text
"transformer"
```

The Decoder Query may attend strongly to the Encoder representation corresponding to:

```text
"transformer"
```

Similarly, while generating:

```text
"Hum"
```

the Decoder may attend to the relevant source representation corresponding to:

```text
"We"
```

Therefore:

> Cross-Attention helps the Decoder align the generated target tokens with relevant information from the source sentence.

---

# ➕ **Add & Norm in Decoder**

The Decoder also uses:

```text
Add

↓

Residual Connection


Norm

↓

Layer Normalization
```

A Decoder layer contains three major sub-layers:

```text
1. Masked Multi-Head Self-Attention

2. Cross-Attention

3. Feed Forward Network
```

Each sub-layer is followed by a Residual Connection and Layer Normalization.

---

# 🔁 **Complete Decoder Layer**

The complete Decoder layer can be represented as:

```text
6 Decoder Layers
```

```text
Target Embedding
        +
Positional Encoding

        ↓

Decoder Layer 1

        ↓

Decoder Layer 2

        ↓

Decoder Layer 3

        ↓

Decoder Layer 4

        ↓

Decoder Layer 5

        ↓

Decoder Layer 6

        ↓

Decoder Output
```

# 🎯 **Final Linear Layer**

After the final Decoder layer, the output is passed through a Linear Layer.

The Linear Layer maps the Decoder representation to the size of the target vocabulary.

Suppose:

```text
Target Vocabulary Size = 50,000
```

Then:

```text
Decoder Output

        ↓

Linear Layer

        ↓

50,000 Logits
```

Each logit represents the score of a possible target token.

---

# 📊 **Softmax**

The logits are converted into probabilities using Softmax.

```text
Logits

↓

Softmax

↓

Token Probabilities
```

For example:

```text
Hum          → 0.05

transformer  → 0.70

sikh         → 0.10

rahe         → 0.03

hai          → 0.02

...
```

The token with the appropriate probability is selected according to the decoding strategy.

# 🧠 **Training vs Inference**

There is an important difference between **Training** and **Inference**.

---

# 🏋️ **Training**

During training, the complete target sentence is available.

For example:

```text
Target:

Hum transformer sikh rahe hai
```

The Decoder input is shifted by one position:

```text
Decoder Input:

<SoS> Hum transformer sikh rahe hai
```

Target output:

```text
Hum transformer sikh rahe hai <EoS>
```

Therefore, the model learns:

```text
<SoS>       → Hum

Hum         → transformer

transformer → sikh

sikh        → rahe

rahe        → hai

hai         → <EoS>
```

---

# 📚 **Teacher Forcing**

During training, the Decoder is given the **correct previous target tokens** rather than its own previous predictions.

For example:

```text
Decoder Input:

<SoS> Hum transformer sikh rahe
```

Target:

```text
Hum transformer sikh rahe hai
```

Therefore:

```text
<SoS>       → Hum

Hum         → transformer

transformer → sikh

sikh        → rahe

rahe        → hai
```

This training technique is called:

> **Teacher Forcing**

Teacher Forcing makes training more efficient because the model does not have to wait for its own previous prediction at every position.

---

# 🚀 **Inference**

During inference, the model does not have the correct target sentence.

Therefore, it generates one token at a time.

```text
<SoS>

↓

Hum

↓

<SoS> Hum

↓

transformer

↓

<SoS> Hum transformer

↓

sikh

↓

<SoS> Hum transformer sikh

↓

rahe

↓

<SoS> Hum transformer sikh rahe

↓

hai

↓

<SoS> Hum transformer sikh rahe hai

↓

<EoS>
```

The process stops when `<EoS>` is generated.

---

# ⚖️ **Training vs Inference**

| Training | Inference |
|---|---|
| Target sequence is available | Target sequence is not available |
| Uses Teacher Forcing | Generates autoregressively |
| Correct previous tokens are provided | Previous model predictions are used |
| Predictions can be made for all positions in parallel | Tokens are generated one by one |
| Causal masking is still used | Causal masking is used |

# 🚀 **Why Transformer?**

Before the Transformer, sequence-to-sequence models commonly used:

- RNN
- LSTM
- GRU

These models process tokens sequentially.

For example:

```text
Token 1

   ↓

Token 2

   ↓

Token 3

   ↓

Token 4
```

This makes parallel processing difficult.

The Transformer uses Self-Attention to model relationships between tokens while allowing highly parallel computation within each layer.

```text
Token 1 ─┐
Token 2 ─┤
Token 3 ─┼──→ Self-Attention
Token 4 ─┤
Token 5 ─┘
```

This made the Transformer highly suitable for large-scale training.

---

# 🧠 **Key Idea of Transformer**

The central idea of the Transformer is:

> **Use Attention to model relationships between tokens instead of relying on recurrent processing.**

The Encoder:

> **Understands the source sentence.**

The Decoder:

> **Generates the target sentence.**

Therefore:

```text
Source Sentence

"We are learning transformer"

        ↓

      Encoder

        ↓

Contextual Representations

        ↓

      Decoder

        ↓

Target Sentence

"Hum transformer sikh rahe hai"
```

# ✅ **Advantages of Transformer**

- **Parallel Training** — We can train it parallely.
- **Long-Term Dependencies** — It can capture long-term dependencies effectively.
- **Better Understanding of Overall Context** — Self-Attention helps the model understand the overall context.
- **Scalable** — Transformer architectures scale effectively with larger models and datasets.
- **Flexible** — The Transformer architecture can be adapted to different NLP tasks and model designs.

---

# ⚠️ **Disadvantages of Transformer**

- **Complex Architecture** — The Transformer architecture is relatively complex.
- **More Computational Resources Required** — Transformers require significant computational resources, especially for large models.
- **More Training Data Needed** — Large Transformer models generally benefit from large amounts of training data.
- **Less Interpretable** — Understanding exactly why a Transformer produces a particular output can be difficult.

# 📌 **Summary**

- The **Transformer** was introduced in **2017** in the paper *Attention Is All You Need*.
- The Transformer is based on **Attention** rather than recurrent processing.
- The original Transformer consists of an **Encoder** and a **Decoder**.
- The Encoder processes the **source language**.
- The Decoder generates the **target language**.
- Input tokens are converted into vectors using an **Embedding Layer**.
- The original Transformer uses an embedding dimension of **512**.
- **Positional Encoding** provides information about the order of tokens.
- The original Transformer uses **sinusoidal positional encoding**.
- The Encoder uses **Multi-Head Self-Attention**.
- Every attention head has its own learnable Query, Key and Value projection matrices.
- Multiple attention heads allow the model to learn different relationships between tokens.
- The outputs of all attention heads are concatenated and passed through an output projection.
- **Residual Connections** help preserve information and improve gradient flow.
- **Layer Normalization** helps stabilize training.
- The **Feed Forward Network** applies a non-linear transformation independently to every token.
- The original Transformer has **6 Encoder layers**.
- The original Transformer has **6 Decoder layers**.
- The original Transformer uses **8 attention heads**.
- The original Transformer uses $d_{model}=512$.
- The original Transformer uses $d_{ff}=2048$.
- The Decoder uses **Masked Self-Attention** to prevent access to future target tokens.
- **Cross-Attention** connects the Decoder with the Encoder.
- In Cross-Attention, **Query comes from the Decoder**, while **Key and Value come from the Encoder**.
- The Decoder generates tokens **autoregressively** during inference.
- `<SoS>` represents the beginning of the target sequence.
- `<EoS>` represents the end of the target sequence.
- During training, **Teacher Forcing** provides the correct previous target tokens.
- During inference, the Decoder uses its own previously generated tokens.
- A **Linear Layer + Softmax** converts Decoder representations into probabilities over the target vocabulary.
- The Transformer became the foundation for many modern architectures such as **BERT, GPT, T5, and other Transformer-based models**.
